# Sorting: Merge Sort

## LeetCode 148 — Sort List

Merge Sort is a sorting algorithm based on **divide and conquer**.

Main ideas:

1. Divide the data into smaller parts.
2. Sort the smaller parts.
3. Merge the sorted parts.
4. Repeat until the complete data is sorted.

Time complexity: **O(n log n)**

Problem:
https://leetcode.com/problems/sort-list/


# 1. What is Sorting?

Sorting means arranging data in an order.

Example:

```text
Before:
[7, 2, 9, 1, 5]

After:
[1, 2, 5, 7, 9]
```

Usually, when we say sorting, we mean arranging from smallest to largest.

We can also sort from largest to smallest.


# 2. What is Merge Sort?

Merge Sort is a sorting method that does not try to sort everything at once.

It follows this idea:

```text
Divide
  ↓
Sort smaller parts
  ↓
Merge
  ↓
Get the final sorted result
```

It is called **divide and conquer** because a large problem is divided into smaller problems, those smaller problems are solved, and their answers are combined.


# 3. Divide and Conquer

Imagine you have 8 books that need to be arranged from shortest to tallest.

Instead of arranging all 8 at once:

```text
8 books
```

divide them:

```text
4 books + 4 books
```

Then divide again:

```text
2 + 2 + 2 + 2
```

Then:

```text
1 + 1 + 1 + 1 + 1 + 1 + 1 + 1
```

A group containing one item is already sorted.

Now we combine the small sorted groups.

That is the basic idea of Merge Sort.


# 4. Simple Merge Sort Example

Start with:

```text
[8, 3, 5, 4, 7, 6, 1, 2]
```

Divide:

```text
[8, 3, 5, 4]    [7, 6, 1, 2]
```

Divide again:

```text
[8, 3] [5, 4]    [7, 6] [1, 2]
```

Divide again:

```text
[8] [3] [5] [4] [7] [6] [1] [2]
```

Now every piece has one element, so each piece is already sorted.

The next job is to merge them.


# 5. What is Merging?

Merging means combining two already-sorted groups into one sorted group.

Example:

```text
[3, 8]
[4, 5]
```

Compare the first values:

```text
3 vs 4 → take 3
8 vs 4 → take 4
8 vs 5 → take 5
```

Then `8` remains.

Result:

```text
[3, 4, 5, 8]
```

The important rule is:

**Always compare the smallest remaining value from the two sorted groups.**

Right side:

```text

```

Step by step:

```text
2 vs 1 → take 1
2 vs 4 → take 2
6 vs 4 → take 4
6 vs 7 → take 6
9 vs 7 → take 7
```

Then `9` remains.

Final result:

```text
[1, 2, 4, 6, 7, 9]
```


# 7. Complete Merge Sort Picture

Starting data:

```text
[8, 3, 5, 4, 7, 6, 1, 2]
```

Split:

```text
[8, 3, 5, 4]    [7, 6, 1, 2]
```

Split:

```text
[8, 3] [5, 4]    [7, 6] [1, 2]
```

Split:

```text
[8] [3] [5] [4] [7] [6] [1] [2]
```

Merge:

```text
[3, 8] [4, 5]    [6, 7] [1, 2]
```

Merge:

```text

```

Final merge:

```text
[1, 2, 3, 4, 5, 6, 7, 8]
```

So the mental model is:

```text
Divide → Divide → Divide → Merge → Merge → Merge
```


# 8. Merge Sort on a Normal Python List

Before working with a linked list, it is helpful to see the basic idea using a normal Python list.

For:

```text
[5, 2, 8, 1]
```

we get:

```text
[5, 2]   [8, 1]
```

then:

```text
[5] [2] [8] [1]
```

then:

```text
[2, 5]   [1, 8]
```

and finally:

```text
[1, 2, 5, 8]
```


````python
def merge_sort(arr):
    # A list with 0 or 1 item is already sorted
    if len(arr) <= 1:
        return arr

    # Divide into two halves
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Merge the sorted halves
    return merge(left, right)


def merge(left, right):
    result = []
    j = 0
    i = 0


    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


nums = [8, 3, 5, 4, 7, 6, 1, 2]

print("Original:", nums)
print("Sorted:  ", merge_sort(nums))
````

# 9. Why is Merge Sort O(n log n)?

Suppose there are `n` items.

Merge Sort keeps dividing the data roughly in half.

For example:

```text
8
↓
4
↓
2
↓
1
```

The number of levels is about:

```text
log₂(n)
```

At every level, the merge process looks at all `n` items.

So:

```text
Work at each level = O(n)
Number of levels   = O(log n)
```

Therefore:

```text
O(n) × O(log n) = O(n log n)
```

So Merge Sort runs in:

```text
O(n log n)
```


# 10. Why O(n log n) is Useful

A common slower sorting approach can take:

```text
O(n²)
```

Merge Sort takes:

```text
O(n log n)
```

For a large amount of data, `n log n` grows much more slowly than `n²`.

That is why Merge Sort is considered an efficient sorting algorithm.

# 12. What is a Linked List?

A linked list is made of nodes.

Each node contains:

```text
value
+
link to the next node
```

Example:

```text
4 → 2 → 1 → 3 → None
```

Think of train carriages.

Each carriage stores information and has a connection to the next carriage.

So:

```text
Node → Node → Node → Node
```

In this problem, we need to rearrange these connected nodes into sorted order.


# 13. Why Merge Sort Works Well for a Linked List

Merge Sort is a good fit because:

- We can split the linked list into smaller lists.
- We can recursively sort those smaller lists.
- We can merge two sorted linked lists by changing their links.
- We do not need to repeatedly access random positions like we would in an array.

So the same pattern works:

```text
Split → Sort → Merge
```


# 14. Finding the Middle of a Linked List

A common technique is the **slow and fast pointer** method.

We keep two pointers:

```text
slow
fast
```

`slow` moves one node at a time.

`fast` moves two nodes at a time.

Example:

```text
1 → 2 → 3 → 4 → 5
```

The fast pointer moves faster, so when it reaches the end, the slow pointer is near the middle.

This lets us split the linked list into two smaller lists.

# 17. The Dummy Node Idea

When merging two linked lists, a temporary dummy node makes the process easier.

Think of it as an empty starting point:

```text
dummy → ...
```

We keep attaching the smaller node to the end.

When finished, the actual sorted list starts after the dummy node.

This is a very common linked-list technique.


# 18. Full Approach for LeetCode 148

The algorithm is:

```text
1. If the list has 0 or 1 node, it is already sorted.
2. Find the middle using slow and fast pointers.
3. Split the list into two halves.
4. Recursively sort the left half.
5. Recursively sort the right half.
6. Merge the two sorted halves.
7. Return the merged result.
```

The full pattern is:

```text
Linked List
    ↓
Find Middle
    ↓
Split
    ↓
Sort Left
    ↓
Sort Right
    ↓
Merge
    ↓
Sorted Linked List
```


````python
# LeetCode 148 - Sort List
# Merge Sort for a linked list

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def sortList(self, head):
        # Base case
        if head is None or head.next is None:
            return head

        # Find the middle using slow/fast pointers
        slow = head
        fast = head.next

        while fast is not None and fast.next is not None:
            slow = slow.next
            fast = fast.next.next

        # Split into two lists
        right = slow.next
        slow.next = None

        # Recursively sort both halves
        left = self.sortList(head)
        right = self.sortList(right)

        # Merge the two sorted halves
        return self.merge(left, right)

    def merge(self, left, right):
        dummy = ListNode()
        current = dummy

        while left is not None and right is not None:
            if left.val <= right.val:
                current.next = left
                left = left.next
            else:
                current.next = right
                right = right.next

            current = current.next

        # Attach any remaining nodes
        if left is not None:
            current.next = left

        if right is not None:
            current.next = right

        return dummy.next
````

# 19. Understanding the Solution

The solution has two important functions.

## `sortList`

Its job is:

```text
Find middle
Split
Sort left
Sort right
Merge
```

## `merge`

Its job is:

```text
Take two sorted linked lists
Compare their current nodes
Attach the smaller node
Continue until one list ends
Attach the remaining nodes
```

Keeping these responsibilities separate makes the solution easier to understand.


# 20. Base Case

The code stops when:

```text
There is no node
or
there is only one node
```

A one-node list is already sorted.

Example:

```text
7
```

There is nothing to do.

This is the stopping condition for the recursion.


# 21. Splitting the List

Suppose:

```text
1 → 2 → 3 → 4
```

The slow pointer moves one step and the fast pointer moves two steps.

Eventually, the slow pointer reaches the position used to divide the list.

We then break the link between the two halves.

Conceptually:

```text
1 → 2     3 → 4
```

This is important because the recursive calls need two separate lists.


# 22. Why We Break the Link

Suppose the original list is:

```text
1 → 2 → 3 → 4
```

If we decide that `2` is the end of the left half, we need:

```text
1 → 2 → None
```

and:

```text
3 → 4 → None
```

Otherwise the two "halves" would still be connected and the recursive process would not work correctly.

So the middle link is cut.

# 24. Testing the Solution

The helper functions below create a linked list and turn the final linked list back into a normal Python list so we can easily see the result.


````python
def build_linked_list(values):
    dummy = ListNode()
    current = dummy

    for value in values:
        current.next = ListNode(value)
        current = current.next

    return dummy.next


def linked_list_to_list(head):
    result = []

    while head is not None:
        result.append(head.val)
        head = head.next

    return result


# Test
head = build_linked_list([4, 2, 1, 3])

solution = Solution()
sorted_head = solution.sortList(head)

print("Input:  [4, 2, 1, 3]")
print("Output:", linked_list_to_list(sorted_head))
````

````python
# More test cases

test_cases = [
    [4, 2, 1, 3],
    [-1, 5, 3, 4, 0],
    [2, 1],
    [1],
    [],
    [3, 3, 1, 2, 3]
]

solution = Solution()

for values in test_cases:
    head = build_linked_list(values)
    sorted_head = solution.sortList(head)

    print("Input: ", values)
    print("Output:", linked_list_to_list(sorted_head))
    print()
````

# 25. Complexity

For `n` nodes:

At each level, the algorithm processes all `n` nodes.

There are about:

```text
log n
```

levels.

Therefore:

```text
Time Complexity = O(n log n)
```

The recursive calls use stack space:

```text
Space Complexity = O(log n)
```

The linked-list nodes themselves are already provided, and the merge process can rearrange their links instead of creating a new array containing all values.


# 26. Common Mistakes

## Forgetting the base case

A list with 0 or 1 node is already sorted.

## Finding the middle but not splitting

The connection between the two halves must be removed.

## Comparing the wrong values

Compare:

```text
left.val
right.val
```

not the node objects themselves.

## Forgetting the remaining nodes

When one list becomes empty, the other list may still contain nodes.

Those remaining nodes must be connected to the result.

## Confusing values and links

A linked list node has:

```text
val
next
```

`val` tells us the number.

`next` tells us which node comes after it.


# 27. Merge Sort vs O(n²) Sorting

A basic sorting method may repeatedly compare many pairs of elements and can take:

```text
O(n²)
```

Merge Sort reduces the problem using divide and conquer:

```text
Divide
↓
Smaller problems
↓
Sort
↓
Merge
```

Its time complexity is:

```text
O(n log n)
```

This is why Merge Sort is much more suitable for large inputs.
