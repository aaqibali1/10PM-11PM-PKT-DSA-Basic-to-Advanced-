## 1. Queue

A **Queue** stores items in an order where the item that comes first is removed first.

This rule is called:

**FIFO = First In, First Out**

Think about people waiting at a ticket counter:

```text
First person → Second person → Third person → Fourth person
     ↓
   served first
```

The first person who joins the line gets served first.

````python
# Simple queue using collections.deque

from collections import deque

q = deque()

# Add items
q.append("Ali")
q.append("Sara")
q.append("Ahmed")

print(q)
````

### What happened?

We added:

```text
Ali → Sara → Ahmed
```

So the queue is:

```text
Front                         Back
  ↓                            ↓
[Ali, Sara, Ahmed]
```

`Ali` is the first person in the queue, so `Ali` will leave first.

````python
# Remove the first item

first = q.popleft()

print("Removed:", first)
print("Queue:", q)
````

## 2. Enqueue

**Enqueue** means:

> Add an item to the back of the queue.

In Python:

```python
q.append(x)
```

Example:

```text
Before:
[10, 20, 30]

enqueue(40)

After:
[10, 20, 30, 40]
```

````python
from collections import deque

q = deque([10, 20, 30])

q.append(40)   # enqueue

print(q)
````

## 3. Dequeue

**Dequeue** means:

> Remove the item from the front of the queue.

In Python:

```python
q.popleft()
```

Example:

```text
Before:
[10, 20, 30, 40]

dequeue()

After:
[20, 30, 40]
```

````python
q = deque([10, 20, 30, 40])

removed = q.popleft()

print("Removed:", removed)
print("Queue:", q)
````

## 4. Important Queue Operations

| Operation | Python | Meaning |
|---|---|---|
| Enqueue | `append(x)` | Add to the back |
| Dequeue | `popleft()` | Remove from the front |
| Front | `q[0]` | Look at the first item |
| Size | `len(q)` | Number of items |
| Empty check | `len(q) == 0` | Check whether queue is empty |

A useful memory trick:

```text
append()   → enter from the back
popleft()  → leave from the front
```

````python
from collections import deque

q = deque()

q.append(10)
q.append(20)
q.append(30)

print("Queue:", q)
print("Front:", q[0])
print("Size:", len(q))
print("Empty:", len(q) == 0)

q.popleft()

print("After dequeue:", q)
````

## 5. Why use `deque` instead of a Python list?

You may see this:

```python
q = []
q.append(10)
q.pop(0)
```

It works, but removing the first item from a list can be slow because the remaining items may need to be shifted.

For queue problems, prefer:

```python
from collections import deque

q = deque()
q.append(x)
q.popleft()
```

This is the standard and efficient approach.

## 6. What is a Deque?

A **Deque** means **Double-Ended Queue**.

A normal queue mainly works like this:

```text
add → [10, 20, 30] → remove
```

A deque lets us add and remove from **both ends**:

```text
      add/remove
          ↓
[10, 20, 30, 40]
          ↑
      add/remove
```

For example, imagine a row of boxes where you can place a new box at either end and also take a box from either end.

````python
from collections import deque

dq = deque([10, 20, 30])

# Add to the right
dq.append(40)

# Add to the left
dq.appendleft(5)

print("Deque:", dq)
````

````python
# Remove from the left
left_item = dq.popleft()

# Remove from the right
right_item = dq.pop()

print("Removed from left:", left_item)
print("Removed from right:", right_item)
print("Deque now:", dq)
````

## 7. Deque Operations

| Operation | Python | Meaning |
|---|---|---|
| Add right | `append(x)` | Add to the back |
| Add left | `appendleft(x)` | Add to the front |
| Remove right | `pop()` | Remove from the back |
| Remove left | `popleft()` | Remove from the front |

Memory trick:

```text
append       → right
appendleft   → left
pop          → right
popleft      → left
```

## 8. Queue vs Deque

### Queue

```text
Add → [ A B C ] → Remove
```

It follows FIFO.

### Deque

```text
Add/Remove ↔ [ A B C ] ↔ Add/Remove
```

Both ends are available.

So:

```text
Queue = one main entry side + one main exit side
Deque  = both sides can be used
```

# LeetCode 933 — Number of Recent Calls

Problem:

https://leetcode.com/problems/number-of-recent-calls/

We receive requests at times `t`.

For every request, count how many requests happened in the last **3000 milliseconds**, including the current request.

The time range is:

```text
[t - 3000, t]
```

Both ends are included.

## 10. Why is a Queue a Good Fit?

Requests arrive in increasing time order.

For example:

```text
[1, 100, 3001, 3002]
 ↑                    ↑
oldest              newest
```

The oldest request is always at the front.

When it becomes too old, remove it from the front.

That gives us exactly:

```text
new request  → enqueue
old request  → dequeue
```

So this problem is a classic:

**Queue + Sliding Window**

## 11. The Main Idea

For every new time `t`:

### Step 1
Add `t` to the queue.

```python
q.append(t)
```

### Step 2
Remove every time that is older than `t - 3000`.

```python
while q[0] < t - 3000:
    q.popleft()
```

### Step 3
The number of remaining requests is the answer.

```python
len(q)
```

````python
from collections import deque

class RecentCounter:

    def __init__(self):
        self.q = deque()

    def ping(self, t):
        # 1. Add the new request
        self.q.append(t)

        # 2. Remove requests that are too old
        while self.q[0] < t - 3000:
            self.q.popleft()

        # 3. Return the number of recent requests
        return len(self.q)
````

This version prints the queue after every call so you can see exactly what is happening.

````python
from collections import deque

class RecentCounter:

    def __init__(self):
        self.q = deque()

    def ping(self, t):
        self.q.append(t)

        while self.q[0] < t - 3000:
            self.q.popleft()

        print(f"t = {t:4} | queue = {list(self.q)} | answer = {len(self.q)}")

        return len(self.q)


counter = RecentCounter()

counter.ping(1)
counter.ping(100)
counter.ping(3001)
counter.ping(3002)
````

## 15. Why do we use `while`?

We use:

```python
while q[0] < t - 3000:
    q.popleft()
```

instead of only:

```python
if q[0] < t - 3000:
    q.popleft()
```

because several old requests may need to be removed.

Example:

```text
Queue:
[1, 10, 50, 100, 5000]

Current t = 5000

Window starts at:
2000
```

The first four values are too old:

```text
1     ❌
10    ❌
50    ❌
100   ❌
5000  ✅
```

`while` keeps removing until the front is valid.

## 16. Why do we only check the front?

Suppose:

```text
[100, 200, 500, 900]
```

The numbers are already sorted.

If `100` is valid, then every later value (`200`, `500`, `900`) is also valid.

If `100` is too old, remove it and check the next oldest value.

Therefore:

```python
q[0]
```

is enough to decide what should leave the queue.

## 17. Brute Force Version

One simple approach is to save every request and check all of them each time.

```python
requests = []

def ping(t):
    requests.append(t)

    count = 0

    for x in requests:
        if t - 3000 <= x <= t:
            count += 1

    return count
```

This is easy to understand, but it may repeatedly scan many old requests.

The queue solution is better because old requests are removed once they are no longer useful.

````python
# Brute force demonstration

class RecentCounterBruteForce:

    def __init__(self):
        self.requests = []

    def ping(self, t):
        self.requests.append(t)

        count = 0

        for x in self.requests:
            if t - 3000 <= x <= t:
                count += 1

        return count


counter = RecentCounterBruteForce()

print(counter.ping(1))
print(counter.ping(100))
print(counter.ping(3001))
print(counter.ping(3002))
````

## 18. Optimized Version

The optimized queue solution is:

```python
from collections import deque

class RecentCounter:

    def __init__(self):
        self.q = deque()

    def ping(self, t):
        self.q.append(t)

        while self.q[0] < t - 3000:
            self.q.popleft()

        return len(self.q)
```

The important pattern is:

```text
Append new item
      ↓
Remove old items from front
      ↓
Count remaining items
```

## 19. Complexity

For each request:

```text
append → O(1)
popleft → O(1)
```

Each request is added once and removed at most once.

So the queue solution is:

**Amortized time: O(1) per `ping()`**

Space:

**O(n)**

where `n` is the number of requests currently stored in the 3000 ms window.
