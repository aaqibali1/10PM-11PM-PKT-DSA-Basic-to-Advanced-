# 1. Tree Recursion

A tree is naturally recursive because every child can be the root of a smaller subtree.

```text
        1
       / \\
      2   3
     / \\
    4   5
```

A common pattern is:

```text
if node is None:
    return base_answer

left_answer = solve(node.left)
right_answer = solve(node.right)

return combine(left_answer, right_answer, node)
```

The key questions are: What is the base case? What should the left and right subtrees return? How do we combine them?

# 2. Height, Depth, and Subtree Reasoning

Depth is the distance from the root to a node.

```text
        1       depth 0
       / \\
      2   3     depth 1
     /
    4           depth 2
```

LeetCode 104 defines maximum depth as the number of nodes on the longest root-to-leaf path.

For a recursive tree problem, think:

```text
current node
├── left subtree
└── right subtree
```

Ask each subtree for its answer and combine those answers at the parent.

# 3. LeetCode 104 — Maximum Depth

Example:

```text
        3
       / \\
      9   20
         /  \\
        15   7
```

The longest path contains 3 nodes.

```text
3 → 20 → 15
```

So the answer is `3`.

The recurrence is:

```text
None → 0
node → 1 + max(left_depth, right_depth)
```


````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def maxDepth(self, root):
        if root is None:
            return 0

        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)

        return 1 + max(left_depth, right_depth)


root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

print(Solution().maxDepth(root))
````

# 4. BST Invariant

A Binary Search Tree satisfies:

```text
all left-subtree values < node < all right-subtree values
```

This rule lets us choose a direction during search and insertion.

```text
target < node → left
target > node → right
target == node → found
```

The invariant must hold for every node, not only direct children.

# 5. BST Insertion

Start:

```text
        8
       / \\
      3   10
     / \\
    1   6
```

Insert `7`:

```text
7 < 8 → left
7 > 3 → right
7 > 6 → right
```

Result:

```text
        8
       / \\
      3   10
     / \\
    1   6
         \\
          7
```


````python
def insert_bst(root, val):
    if root is None:
        return TreeNode(val)

    if val < root.val:
        root.left = insert_bst(root.left, val)
    elif val > root.val:
        root.right = insert_bst(root.right, val)

    return root
````

# 6. BST Minimum, Maximum, Predecessor, Successor

```text
Minimum     → farthest left
Maximum     → farthest right
Predecessor → largest value smaller than current
Successor   → smallest value larger than current
```

If a node has a left subtree, its predecessor is the rightmost node of that subtree.

If a node has a right subtree, its successor is the leftmost node of that subtree.

These ideas become especially important in BST deletion.

# 7. BST Deletion

There are three cases.

## Case 1: No children

Remove the node.

## Case 2: One child

Replace the node with its only child.

## Case 3: Two children

Replace the node with either:

```text
inorder successor = smallest value in right subtree
```

or:

```text
inorder predecessor = largest value in left subtree
```

Then delete that replacement from its original position.

# 8. Why Two-Child Deletion Works

Suppose we delete `8`:

```text
        8
       / \\
      3   12
     / \\
    1   6
       / \\
      4   7
```

The successor of `8` is the smallest value in the right subtree, which is `12` here.

The predecessor is the largest value in the left subtree, which is `7`.

Either can replace `8` while preserving the BST ordering.

# 9. LeetCode 98 — Validate Binary Search Tree

Checking only parent-child pairs is not enough.

This is invalid:

```text
        8
       / \\
      3   10
         /
        5
```

`5 < 10`, but `5` is in the right subtree of `8`, so it must also be `> 8`.

The key idea is to carry an allowed range from ancestors.

# 10. Range-Based Validation

Start at the root with:

```text
(-∞, +∞)
```

For a node `x`:

```text
left subtree  → (low, x)
right subtree → (x, high)
```

The range gets narrower as we go down the tree.

For the invalid `5` above:

```text
5 must be in (8, 10)
```

It is not, so the tree is invalid.

````python
# LeetCode 98 - Validate Binary Search Tree

class Solution:
    def isValidBST(self, root):
        def validate(node, low, high):
            if node is None:
                return True

            if not (low < node.val < high):
                return False

            return (
                validate(node.left, low, node.val)
                and
                validate(node.right, node.val, high)
            )

        return validate(root, float("-inf"), float("inf"))
````

# 11. Validate BST Dry Run

Valid tree:

```text
        8
       / \\
      3   10
     / \\
    1   6
```

Ranges:

```text
8 → (-∞, +∞)
3 → (-∞, 8)
1 → (-∞, 3)
6 → (3, 8)
10 → (8, +∞)
```

Every node satisfies its range, so the answer is `True`.

This method checks the full ancestor constraint, not just the parent.

````python
# Test LeetCode 98

valid = TreeNode(2)
valid.left = TreeNode(1)
valid.right = TreeNode(3)

invalid = TreeNode(5)
invalid.left = TreeNode(1)
invalid.right = TreeNode(4)
invalid.right.left = TreeNode(3)
invalid.right.right = TreeNode(6)

print("Valid tree:", Solution().isValidBST(valid))
print("Invalid tree:", Solution().isValidBST(invalid))
````

# 12. Inorder Validation Idea

Another BST property:

> Inorder traversal of a valid BST is strictly increasing when duplicates are not allowed.

Example:

```text
        4
       / \\
      2   7
     / \\
    1   3
```

Inorder:

```text
1, 2, 3, 4, 7
```

This gives another useful way to recognize a valid BST.

# 13. Complexity

For a BST with height `h`:

```text
Search    → O(h)
Insert    → O(h)
Delete    → O(h)
Min/Max   → O(h)
```

Balanced tree:

```text
h = O(log n)
```

Highly skewed tree:

```text
h = O(n)
```

Validation visits all `n` nodes:

```text
O(n) time
```

Recursive space is proportional to the tree height.

# 14. Common Mistakes

- Forgetting the `None` base case in recursion.
- Comparing only immediate parent and child when validating a BST.
- Forgetting to reconnect a modified subtree after recursive insertion/deletion.
- Mixing up predecessor and successor.
- Assuming every BST operation is `O(log n)`; this depends on the tree height.

Memory tricks:

```text
Minimum     → leftmost
Maximum     → rightmost
Predecessor → largest smaller
Successor   → smallest larger
```

For tree recursion:

```text
Base case → solve left → solve right → combine
```

For BST validation:

```text
Carry the allowed (low, high) range downward.
```
