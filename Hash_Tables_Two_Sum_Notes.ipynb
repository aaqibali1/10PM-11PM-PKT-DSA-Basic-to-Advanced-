# Hash Tables

## 1. What is a Hash Table?

A hash table is a way to store information so that we can find it very quickly.

It stores information in the form:

**key → value**

Think about a phone contacts list.

You know a person's name and want to find their phone number.

For example:

Name → Phone Number

Ali → 0300-1234567  
Sara → 0311-7654321  
Ahmed → 0322-5555555

Here:

- The name is the **key**
- The phone number is the **value**

Instead of checking every contact one by one, a hash table is designed to find the value very quickly using the key.


# 2. Key-Value Pair

A hash table stores information as pairs.

```text
Key       → Value

"Ali"     → "0300-1234567"
"Sara"    → "0311-7654321"
"Ahmed"   → "0322-5555555"
```

Suppose you ask:

> What is Sara's phone number?

The hash table uses the key `Sara` to find the corresponding value.

```text
Sara → 0311-7654321
```

This is called a **key-value lookup**.

The main idea is:

**Use a key to quickly find its value.**


# 3. Hashing

Hashing is the process of converting a key into a location where its information can be stored.

You do not need to think about the mathematical details at first.

Think of it like a smart filing system.

Suppose a library has thousands of books.

Instead of looking through every book, the library gives each book a location based on some rule.

```text
Book name → Rule → Shelf location
```

A hash table does something similar:

```text
Key → Hashing process → Storage location
```

This helps us find information quickly.


# 4. Why Hashing is Useful

Imagine you have 1,000 student records.

You want to find the record for:

```text
Student ID = 1052
```

Without a good lookup system, you might check records one by one.

With a hash table, the ID can be used to quickly determine where the record should be.

So instead of thinking:

```text
Check 1
Check 2
Check 3
Check 4
...
```

you can think:

```text
Student ID → quickly locate the record
```

That is why hash tables are extremely useful in DSA.


# 5. Key-Value Lookup

A lookup means:

> "I know the key. Find me its value."

Example:

```text
Username → Email

james → james@example.com
sara  → sara@example.com
ali   → ali@example.com
```

If we know:

```text
sara
```

we can look up:

```text
sara → sara@example.com
```

This is a **key-value lookup**.


# 6. Frequency Map

A very common use of a hash table is counting how many times something appears.

This is called a **frequency map**.

Suppose we have:

```text
Apple, Banana, Apple, Orange, Banana, Apple
```

We can count them:

```text
Apple   → 3
Banana  → 2
Orange  → 1
```

The item is the key.

The number of occurrences is the value.

```text
Item → Frequency
```

So a frequency map is simply:

**"How many times did each thing appear?"**


# 7. Real-Life Frequency Example

Imagine a teacher asks students which fruit they like.

The answers are:

```text
Apple
Banana
Apple
Mango
Banana
Apple
```

The teacher wants to know how many students chose each fruit.

The result is:

```text
Apple  → 3
Banana → 2
Mango  → 1
```

A hash table is very useful for this type of counting problem.

This pattern appears frequently in:

- Counting characters
- Counting numbers
- Counting words
- Finding duplicates
- Comparing frequencies


# 8. What is a Collision?

A collision happens when two different keys are sent to the same storage location by the hashing process.

Think about two students being assigned the same locker.

```text
Student A → Locker 5
Student B → Locker 5
```

Both cannot simply occupy the same locker in the same way.

The hash table therefore needs a way to handle this situation.

This is called **collision handling**.


# 9. Why Can Collisions Happen?

Imagine there are only 5 available lockers:

```text
Locker 1
Locker 2
Locker 3
Locker 4
Locker 5
```

But you have many students.

Eventually, two students may be assigned to the same locker number.

The same basic idea applies to hash tables.

There may be many possible keys but a limited number of storage locations.

So different keys can sometimes lead to the same location.

That is a collision.


# 10. Collision Handling

A hash table needs a strategy for dealing with collisions.

A simple way to imagine one strategy is:

```text
Location 5:
Student A
Student B
Student C
```

The table keeps the items connected at that location and can still find the correct one.

The exact implementation can vary, but the important idea for DSA is:

**Different keys can sometimes map to the same location, and the hash table must handle that safely.**


# 11. Hash Table in DSA

When solving DSA problems, a hash table is especially useful when you need to:

- Find something quickly
- Check whether something already exists
- Store a relationship between two things
- Count frequencies
- Remember information seen earlier

A useful question to ask is:

> "Do I need to quickly find something that I have already seen?"

If the answer is yes, a hash table may be useful.


# Two Sum

## 12. The Problem

LeetCode Problem 1 is called **Two Sum**.

https://leetcode.com/problems/two-sum/

You are given a list of numbers and a target number.

You need to find **two different numbers whose sum equals the target**.

For example:

```text
Numbers:
[2, 7, 11, 15]

Target:
9
```

We need two numbers that add up to 9.

```text
2 + 7 = 9
```

So the answer is the positions of `2` and `7`.


# 13. Understanding Two Sum Like a Non-Coder

Imagine you have a box containing numbers:

```text
2, 7, 11, 15
```

Someone says:

> "Find two numbers that together make 9."

You pick `2`.

Then you ask:

> "What number do I need to make 9?"

The answer is:

```text
9 - 2 = 7
```

You look for `7`.

It exists.

So you found the pair:

```text
2 + 7 = 9
```

This idea is the key to Two Sum.


# 14. The Important Idea: Complement

The number you need is called the **complement**.

Suppose the target is:

```text
9
```

and you currently have:

```text
2
```

You need:

```text
9 - 2 = 7
```

So:

```text
2 → needs 7
```

If you see `7`, then you have found the answer.

Another example:

```text
Target = 10
Current number = 6

Needed number:
10 - 6 = 4
```

So:

```text
6 → needs 4
```

The main idea is:

**For every number, calculate the number needed to reach the target.**


# 15. Why Use a Hash Table for Two Sum?

Suppose the list is:

```text
[2, 7, 11, 15]
```

and the target is:

```text
9
```

As we look at each number, we want to quickly answer:

> "Have I already seen the number I need?"

A hash table is very good at this.

For `2`:

```text
Needed = 7
```

We remember what we have seen so far.

Then we reach `7`.

Now we ask:

```text
Have I already seen 7?
```

or, depending on the exact way we think about the process:

```text
Have I already seen the number needed by 7?
```

For `7`, the needed number is:

```text
9 - 7 = 2
```

And `2` was already seen.

Therefore, we found the pair.


# 16. Two Sum Step by Step

Consider:

```text
Numbers = [2, 7, 11, 15]
Target = 9
```

### Step 1: Look at 2

We need:

```text
9 - 2 = 7
```

Have we seen `7` yet?

No.

So we remember `2`.

```text
Seen:
2
```

### Step 2: Look at 7

We need:

```text
9 - 7 = 2
```

Have we seen `2`?

Yes.

So:

```text
2 + 7 = 9
```

We found the answer.

We stop here.


# 17. What Is the Hash Table Storing?

In Two Sum, the hash table can remember numbers we have already seen.

It can also remember their positions.

Think of it like this:

```text
Number → Position

2  → position 0
7  → position 1
11 → position 2
15 → position 3
```

This is useful because the problem does not just ask for the two numbers.

It asks for their **indices**, which means their positions in the list.


# 18. Understanding the Final Answer

For:

```text
Numbers = [2, 7, 11, 15]
Target = 9
```

The pair is:

```text
2 and 7
```

Their positions are:

```text
2 → position 0
7 → position 1
```

So the answer is:

```text
[0, 1]
```

The important distinction is:

```text
Values:
2 and 7

Indices:
0 and 1
```


# 19. Another Two Sum Example

Suppose:

```text
Numbers = [3, 2, 4]
Target = 6
```

Start with `3`.

We need:

```text
6 - 3 = 3
```

We have not found another `3`.

Move to `2`.

We need:

```text
6 - 2 = 4
```

We have not seen `4` yet.

Move to `4`.

We need:

```text
6 - 4 = 2
```

We already saw `2`.

Therefore:

```text
2 + 4 = 6
```

Their positions are:

```text
2 → position 1
4 → position 2
```

Answer:

```text
[1, 2]
```


# 20. Why Not Just Try Every Pair?

A simple approach is to compare every number with every other number.

For example:

```text
2 with 7
2 with 11
2 with 15
7 with 11
7 with 15
11 with 15
```

For a small list, this is fine.

But imagine a list containing thousands or millions of numbers.

Trying every possible pair can become very slow.

A hash table lets us quickly check whether the number we need has already appeared.

That is the main optimization.


# 21. The Pattern to Remember

Two Sum follows this thinking:

```text
Current number
      ↓
What number do I need?
      ↓
Target - current number
      ↓
Have I already seen that number?
      ↓
Yes → Found the pair
No  → Remember the current number
```

This is the important DSA pattern.


# 22. Hash Table + Two Sum

The connection is:

```text
Hash Table
    ↓
Remember numbers already seen
    ↓
Quickly check whether a needed number exists
    ↓
Find the pair
```

This is why **Two Sum** is one of the classic hash table problems.


# 23. Key Concepts to Remember

### Hash Table

A structure for storing and finding information quickly.

```text
Key → Value
```

### Hashing

A process used to determine where information should be stored.

```text
Key → Hashing → Location
```

### Collision

When different keys end up at the same location.

```text
Key A → Location 5
Key B → Location 5
```

### Key-Value Lookup

Use a key to quickly find its value.

```text
Name → Phone Number
```

### Frequency Map

Use a hash table to count how often something appears.

```text
Apple → 3
Banana → 2
Mango → 1
```

### Two Sum

Find two numbers whose sum equals a target.

The key question is:

```text
What number do I need?

Needed = Target - Current Number
```


# 24. Easy Memory Trick

Remember hash tables with these four ideas:

```text
Store
Find
Count
Match
```

**Store** information as key-value pairs.

**Find** values quickly using keys.

**Count** how many times things appear.

**Match** what you have with what you need.

For Two Sum:

```text
Current number + Needed number = Target
```

and:

```text
Needed number = Target - Current number
```

That is the central idea of the problem.
