## 1. What Is Recursion?

**Recursion** is a programming technique in which a function **calls itself** to solve a smaller version of the same problem.

A simple mental model is:

> **Solve a small part, then ask the same function to solve a smaller part.**

For recursion to work correctly, we usually need two things:

1. **Base case** — tells the function when to stop.
2. **Recursive case** — tells the function how to call itself again.

## 2. A Very Simple Recursion Example

Let's start with a countdown.

The function prints the current number and then calls itself with a smaller number.

````python
def countdown(n):
    # Base case: stop when n becomes 0
    if n == 0:
        return

    print(n)

    # Recursive case
    countdown(n - 1)


countdown(7)
````

### What happened?

The calls were:

```text
countdown(5)
    ↓
countdown(4)
    ↓
countdown(3)
    ↓
countdown(2)
    ↓
countdown(1)
    ↓
countdown(0)
```

When `n == 0`, the **base case** is reached and the function stops.

# 3. Base Case

The **base case** is the condition that tells recursion:

> **Stop! Do not call the function again.**

Example:

```python
if n == 0:
    return
```

Here, `n == 0` is the base case.

### Why is the base case important?

Without a base case, the function may keep calling itself indefinitely until Python raises a recursion error.

Example of a bad recursive function:

````python
# DO NOT run this cell.
# It has no proper stopping condition.

# def bad_countdown(n):
#     print(n)
#     bad_countdown(n - 1)
````

### Rule to remember

> **Base case = stopping condition**

# 4. Recursive Case

The **recursive case** is the part where the function calls itself.

In our countdown example:

```python
countdown(n - 1)
```

is the recursive case.

It means:

> “I handled `n`. Now solve the smaller problem `n - 1`.”

So:

```text
Base case       → n == 0
Recursive case  → countdown(n - 1)
```

A useful formula is:

> **Recursion = Base Case + Recursive Case**

# 5. Why Should the Problem Become Smaller?

A recursive call should normally move **closer to the base case**.

Good:

```text
5 → 4 → 3 → 2 → 1 → 0
```

Bad:

```text
5 → 6 → 7 → 8 → 9 → ...
```

The second version moves away from the stopping condition.

### Ask yourself whenever you write recursion:

> **Does each recursive call move me closer to the base case?**

# 6. The Call Stack

The **call stack** is the structure Python uses to keep track of active function calls.

Think of it like a stack of plates:

```text
Plate 3
Plate 2
Plate 1
```

The last plate added is the first one removed.

This is called:

**LIFO — Last In, First Out**

Function calls behave in a similar way.

## Call stack example

For:

```python
countdown(3)
```

the calls grow like this:

```text
countdown(3)
countdown(2)
countdown(1)
countdown(0)
```

At the deepest point, we can visualize the stack as:

```text
┌──────────────┐
│ countdown(0) │
├──────────────┤
│ countdown(1) │
├──────────────┤
│ countdown(2) │
├──────────────┤
│ countdown(3) │
└──────────────┘
```

When `countdown(0)` returns, Python goes back to the previous call.

This is called **unwinding the recursion**.

# 7. Two Phases of Recursion

It is helpful to think of recursion as having two phases.

### Phase 1 — Going down

The function keeps making recursive calls:

```text
3
↓
2
↓
1
↓
0
```

### Phase 2 — Coming back up

After reaching the base case, the calls return:

```text
0
↑
1
↑
2
↑
3
```

So a common recursion pattern is:

```text
GO DOWN
   ↓
   ↓
BASE CASE
   ↑
   ↑
COME BACK UP
```

# 8. What Is Tracing?

**Tracing** means manually following every function call and return.

Tracing is one of the best ways to understand recursion.

For:

```python
countdown(3)
```

trace it as:

```text
countdown(3)
  prints 3
  calls countdown(2)

countdown(2)
  prints 2
  calls countdown(1)

countdown(1)
  prints 1
  calls countdown(0)

countdown(0)
  reaches base case
  returns
```

Output:

```text
3
2
1
```

# 9. Fibonacci Number

Now we can move to **Fibonacci**, which demonstrates a more advanced form of recursion.

LeetCode problem:

## **509. Fibonacci Number**

The Fibonacci sequence begins:

```text
0, 1, 1, 2, 3, 5, 8, 13, 21, ...
```

The mathematical rule is:

```text
F(0) = 0
F(1) = 1

F(n) = F(n - 1) + F(n - 2)
```

Each Fibonacci number is the sum of the previous two numbers.

## Examples

### F(0)

```text
F(0) = 0
```

### F(1)

```text
F(1) = 1
```

### F(2)

```text
F(2) = F(1) + F(0)
     = 1 + 0
     = 1
```

### F(3)

```text
F(3) = F(2) + F(1)
     = 1 + 1
     = 2
```

### F(4)

```text
F(4) = F(3) + F(2)
     = 2 + 1
     = 3
```

### F(5)

```text
F(5) = F(4) + F(3)
     = 3 + 2
     = 5
```

# 10. Recursive Fibonacci Solution

A direct recursive solution is:

```python
def fib(n):
    if n == 0:
        return 0

    if n == 1:
        return 1

    return fib(n - 1) + fib(n - 2)
```

Let's run it.

````python
def fib(n):
    # Base case 1
    if n == 0:
        return 0

    # Base case 2
    if n == 1:
        return 1

    # Recursive case
    return fib(n - 1) + fib(n - 2)


print("fib(0) =", fib(0))
print("fib(1) =", fib(1))
print("fib(2) =", fib(2))
print("fib(3) =", fib(3))
print("fib(4) =", fib(4))
print("fib(5) =", fib(5))
````

# 11. Identify the Base Cases and Recursive Case

Look at the code:

```python
if n == 0:
    return 0

if n == 1:
    return 1

return fib(n - 1) + fib(n - 2)
```

### Base cases

```text
F(0) = 0
F(1) = 1
```

### Recursive case

```text
F(n) = F(n - 1) + F(n - 2)
```

This is **branching recursion** because each non-base call creates **two recursive calls**.

# 12. Trace `fib(4)`

Let's follow the recursive calls.

```text
fib(4)
├── fib(3)
│   ├── fib(2)
│   │   ├── fib(1) → 1
│   │   └── fib(0) → 0
│   │   → 1
│   └── fib(1) → 1
│   → 2
└── fib(2)
    ├── fib(1) → 1
    └── fib(0) → 0
    → 1
```

Therefore:

```text
fib(4) = 2 + 1
       = 3
```

# 13. Understanding the Call Stack in Fibonacci

Suppose we call:

```text
fib(3)
```

The first call is:

```text
fib(3)
```

To finish it, Python first needs:

```text
fib(2)
```

So the stack grows:

```text
fib(2)
fib(3)
```

Then `fib(2)` needs:

```text
fib(1)
```

Now:

```text
fib(1)
fib(2)
fib(3)
```

`fib(1)` is a base case and returns `1`.

Then Python continues with the unfinished work.

This is why understanding the call stack is so important in recursion.

# 14. Why Naive Recursive Fibonacci Is Slow

The simple recursive solution repeats calculations.

For example:

```text
fib(5)
├── fib(4)
│   ├── fib(3)
│   │   ├── fib(2)
│   │   └── fib(1)
│   └── fib(2)
└── fib(3)
    ├── fib(2)
    └── fib(1)
```

Notice that `fib(3)` and `fib(2)` are calculated more than once.

That repeated work makes the naive recursive solution inefficient.

# 15. Complexity Check

For the straightforward recursive Fibonacci implementation:

```python
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)
```

### Time complexity

```text
O(2^n)
```

The number of recursive calls grows exponentially.

### Space complexity

```text
O(n)
```

The deepest chain of recursive calls is about `n` levels.

### Important idea

Even though recursion makes the code look elegant, the **naive Fibonacci recursion is not efficient** because of repeated work.

# 20. Recursion Cheat Sheet

| Concept | Simple Meaning |
|---|---|
| **Recursion** | A function calls itself |
| **Base case** | The stopping condition |
| **Recursive case** | The self-call |
| **Call stack** | Keeps track of active function calls |
| **Tracing** | Following calls step by step |
| **Unwinding** | Returning after the base case |
| **Fibonacci** | Each value uses the previous two |
| **Naive Fibonacci time** | `O(2^n)` |
| **Naive Fibonacci space** | `O(n)` |

```text
        ↓
Make the input smaller
        ↓
Trace the calls
        ↓
Check the call stack
        ↓
Test the solution
        ↓

```

# 21. LeetCode 509 — Fibonacci Number

**Problem:** Return the `n`th Fibonacci number.

**Key definition:**

```text
F(0) = 0
F(1) = 1
F(n) = F(n - 1) + F(n - 2)
```

### Link

https://leetcode.com/problems/fibonacci-number/

### Important teaching note

The recursive Fibonacci solution is excellent for learning recursion because it clearly demonstrates:

- base cases,
- recursive cases,
- branching recursive calls,
- the call stack,
- tracing,
- and complexity analysis.

For production-quality or performance-focused solutions, students should later learn **memoization**, **dynamic programming**, and **iterative Fibonacci**.

````python
# Final runnable version

def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)


# Example tests
test_cases = {
    0: 0,
    1: 1,
    2: 1,
    3: 2,
    4: 3,
    5: 5,
    6: 8,
    7: 13,
}

for n, expected in test_cases.items():
    result = fib(n)
    print(f"fib({n}) = {result} | expected = {expected} | {'PASS' if result == expected else 'FAIL'}")
````
