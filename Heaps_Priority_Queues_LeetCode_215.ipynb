# 1. What is a Heap?

A heap is a special type of complete binary tree.

It keeps one important value at the top.

There are two common types:

```text
Min-Heap → smallest value at the top
Max-Heap → largest value at the top
```

A heap is useful when we repeatedly need the smallest or largest item quickly.


# 2. Complete Binary Tree

A complete binary tree fills each level from left to right.

Example:

```text
        10
       /  \
      20   30
     / \   /
    40 50 60
```

This shape allows a heap to be stored efficiently in an array.


# 3. Min-Heap

A min-heap follows:

```text
Parent <= children
```

Example:

```text
        2
       / \
      5   4
     / \
    9   7
```

The smallest value is always at the root:

```text
2
```

The rest does not have to be completely sorted.


# 4. Max-Heap

A max-heap follows:

```text
Parent >= children
```

Example:

```text
        9
       / \
      7   8
     / \
    2   5
```

The largest value is at the root:

```text
9
```


# 5. Heap is Not a Sorted Array

A heap only guarantees the parent-child relationship.

For example, this is a valid min-heap:

```text
        2
       / \
      5   4
     / \
    9   7
```

Although:

```text
5 > 4
```

the heap is still valid because every parent is smaller than its children.

So:

```text
Heap ≠ fully sorted data
```


# 6. Priority Queue

A normal queue uses:

```text
FIFO
First In, First Out
```

A priority queue removes the item with the highest priority first.

For example, in an emergency room, a more urgent patient may be treated before someone who arrived earlier.

A heap is commonly used to implement a priority queue.


# 7. Heap Operations

Important operations are:

```text
Peek top  → see minimum/maximum
Push      → add an element
Pop       → remove the top element
Heapify   → build/repair a heap
```

Typical complexities:

```text
Peek    → O(1)
Push    → O(log n)
Pop     → O(log n)
Heapify → O(n)
```


# 8. Heap Stored in an Array

A heap can be represented by an ordinary list.

Example:

```text
        2
       / \
      5   4
     / \
    9   7
```

Array:

```text
[2, 5, 4, 9, 7]
```

For an element at index `i`:

```text
left child  = 2*i + 1
right child = 2*i + 2
parent      = (i - 1) // 2
```

This is why we do not need separate tree-node objects for a heap.


# 9. Heapify

Heapify means arranging values so they satisfy the heap property.

Example:

```text
[5, 2, 8, 1, 7]
```

After building a min-heap, the smallest value will be at index `0`.

The entire list does not become sorted; it only becomes a valid heap.


````python
import heapq

nums = [5, 2, 8, 1, 7]

heapq.heapify(nums)

print("Heap:", nums)
print("Minimum:", nums[0])
````

# 10. Push

Adding an element is done with:

```text
heappush(heap, value)
```

The new value is inserted and the heap property is restored.

A new value may move upward. This is often called:

```text
sift up
```

The operation takes:

```text
O(log n)
```


````python
heap = []

heapq.heappush(heap, 5)
heapq.heappush(heap, 2)
heapq.heappush(heap, 8)
heapq.heappush(heap, 1)

print("Heap:", heap)
print("Minimum:", heap[0])
````

# 11. Pop

Removing the top is done with:

```text
heappop(heap)
```

For a min-heap, this removes the smallest value.

After removing it, the heap repairs itself by moving another value downward.

This is often called:

```text
sift down
```

Complexity:

```text
O(log n)
```


````python
heap = [5, 2, 8, 1, 7]
heapq.heapify(heap)

smallest = heapq.heappop(heap)

print("Removed:", smallest)
print("Heap:", heap)
````

# 12. Python `heapq` is a Min-Heap

Python's standard `heapq` module provides a min-heap.

So:

```text
heap[0]
```

is always the smallest element.

For a max-heap, a common trick is to store negative values.


````python
max_heap = []

for value in [10, 5, 8, 12]:
    heapq.heappush(max_heap, -value)

largest = -heapq.heappop(max_heap)

print("Largest:", largest)
````

# 13. Max-Heap Using Negative Values

Suppose the original values are:

```text
[10, 5, 8]
```

Store:

```text
[-10, -5, -8]
```

Because `-10` is the smallest number, `heapq` removes it first.

Convert back:

```text
-(-10) = 10
```

So the largest original value comes out first.


# 14. LeetCode 215 — Kth Largest

The problem asks for the kth largest number.

Example:

```text
nums = [3, 2, 1, 5, 6, 4]
k = 2
```

Sorted:

```text
[1, 2, 3, 4, 5, 6]
```

The 2nd largest is:

```text
5
```


# 15. Main Heap Idea

We only need the:

```text
k largest values
```

So keep a **min-heap of size k**.

Why a min-heap?

Because the smallest of our current k values stays at the top.

That smallest value is the boundary:

```text
values above it → definitely among the top k
value at the top → kth largest
```


# 16. Example with k = 2

Input:

```text
[3, 2, 1, 5, 6, 4]
```

Keep only the largest two.

Process:

```text
3 → [3]
2 → [2, 3]
1 → [2, 3]   because 1 is removed
5 → [3, 5]
6 → [5, 6]
4 → [5, 6]
```

Final heap:

```text
[5, 6]
```

The smallest of these two is:

```text
5
```

Therefore:

```text
2nd largest = 5
```


# 17. Why Remove the Smallest?

Suppose our current top 3 candidates are:

```text
[20, 50, 90]
```

The smallest of these is:

```text
20
```

There are two values larger than it:

```text
50, 90
```

Therefore `20` is the:

```text
3rd largest
```

If a new value is larger than `20`, it deserves a place among the top 3.

So remove `20` and keep the new value.


# 18. Algorithm

For every number:

```text
Push number into min-heap

If heap size > k:
    pop the smallest
```

At the end:

```text
heap[0]
```

is the kth largest element.

Memory trick:

```text
K largest → min-heap of size K
```


````python
# LeetCode 215 - Min-Heap Solution

import heapq


class Solution:
    def findKthLargest(self, nums, k):
        heap = []

        for num in nums:
            heapq.heappush(heap, num)

            if len(heap) > k:
                heapq.heappop(heap)

        return heap[0]
````

# 19. Understand the Code

```text
heap = []
```

Start with an empty heap.

For each number:

```text
heappush()
```

adds it.

Then:

```text
if len(heap) > k:
```

checks whether we have too many candidates.

If yes:

```text
heappop()
```

removes the smallest.

Therefore the heap never keeps more than `k` elements.

# 21. Why `heap[0]` is the Answer

At the end, the heap contains exactly the largest `k` values.

For:

```text
k = 3
```

suppose the heap contains:

```text
[20, 50, 90]
```

The smallest value in this group is:

```text
20
```

There are exactly two values greater than it.

Therefore it is the 3rd largest.

So:

```text
heap[0] = kth largest
```

# 24. Complexity

Let:

```text
n = number of elements
k = requested rank
```

Each element may be pushed:

```text
O(log k)
```

Some elements are also popped:

```text
O(log k)
```

There are `n` elements.

Therefore:

```text
Time = O(n log k)
Space = O(k)
```

The heap never grows beyond `k` elements.


# 25. Heap vs Sorting

A sorting solution would usually take:

```text
O(n log n)
```

The heap solution takes:

```text
O(n log k)
```

When `k` is much smaller than `n`, keeping only `k` candidates can save work and memory.

Quickselect is another approach and has average:

```text
O(n)
```

but the heap method is often easier to reason about.


# 26. Min-Heap vs Max-Heap for Kth Problems

For kth largest:

```text
min-heap of size k
```

For kth smallest:

```text
max-heap of size k
```

Why?

The heap top acts as the boundary of the group we are keeping.


# 27. Common Mistakes

### Mistake 1
Thinking a heap is fully sorted.

It is not.

### Mistake 2
Using a max-heap for the standard size-k kth-largest solution.

The simple size-k approach uses a **min-heap**.

### Mistake 3
Forgetting:

```text
if len(heap) > k:
    pop
```

### Mistake 4
Returning the largest value from the heap.

The top is the **smallest among the largest k values**, which is the kth largest.

### Mistake 5
Forgetting Python `heapq` is a min-heap.

# 30. Final Memory Trick

The most important idea is:

**Use a min-heap to maintain the current top `k` largest elements. The smallest element inside that group is exactly the kth largest element.**

Think:

```text
All numbers
    ↓
Keep only the largest k
    ↓
Min-heap keeps the smallest of those k at the top
    ↓
That top value = kth largest
```
