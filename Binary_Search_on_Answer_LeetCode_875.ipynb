# Binary Search on Answer

## LeetCode 875 — Koko Eating Bananas

Binary Search on Answer is a pattern where we search for the smallest or largest value that satisfies a condition.

Unlike normal Binary Search, we are usually not searching for a target already inside a sorted array.

For Koko, the possible answers are eating speeds.

We ask:

> Can Koko finish all banana piles within the allowed time at this speed?

Problem:
https://leetcode.com/problems/koko-eating-bananas/


# 1. Normal Binary Search vs Binary Search on Answer

Normal Binary Search:

```text
Sorted data
    ↓
Find a target
```

Example:

```text
[10, 20, 30, 40, 50]
```

Find `40`.

Binary Search on Answer:

```text
Possible answers
    ↓
Test one candidate
    ↓
Can it work?
    ↓
Yes or No
```

For Koko:

```text
speed = 1, 2, 3, 4, 5, ...
```

We want the smallest speed that works.


# 2. What is a Feasibility Check?

A feasibility check is simply a Yes/No question.

For Koko:

> Can she finish all bananas within `h` hours at speed `k`?

For example:

```text
speed 1 → No
speed 2 → No
speed 3 → No
speed 4 → Yes
speed 5 → Yes
speed 6 → Yes
```

This is useful because the results are monotonic.

Once a speed works, every larger speed also works.


# 3. Monotonic Predicate

A predicate is a Yes/No condition.

A monotonic predicate changes in one direction.

For Koko:

```text
No  No  No  No  Yes  Yes  Yes  Yes
                  ↑
            first feasible speed
```

This lets us use Binary Search to find the boundary.

The important pattern is:

```text
Impossible → Impossible → Possible → Possible
```

We want the first possible value.


# 4. Real-Life Example

Imagine a delivery truck.

You want the minimum truck capacity that can finish all deliveries within 5 days.

Possible capacities:

```text
10 → No
20 → No
30 → No
40 → Yes
50 → Yes
60 → Yes
```

The answer is `40`.

You do not need to test every possible capacity.

You can test a middle capacity and eliminate half of the possibilities.

This is Binary Search on Answer.


# 5. What is LeetCode 875?

Koko has several piles of bananas.

Example:

```text
piles = [3, 6, 7, 11]
```

She has:

```text
h = 8 hours
```

She chooses an integer eating speed `k`.

During one hour, she eats at most `k` bananas from one pile.

If the pile has fewer than `k` bananas, she finishes that pile and does not start another pile during the same hour.

The goal:

> Find the minimum integer speed that allows her to finish every pile within `h` hours.


```text
ceil(11 / 3) = 4
```



# 8. Finding the Search Space

The minimum speed is:

```text

```


For:

```text
[3, 6, 7, 11]
```

```text

```

We do not need speeds larger than 11 because at speed 11 the largest pile can be finished in one hour.

# 10. The Core Binary Search Rule

At every step:

```text
Candidate speed
      ↓
Calculate required hours
      ↓
Is required hours <= h?
```

If Yes:

```text
The speed works.
Try smaller speeds.
```

If No:

```text
The speed is too slow.
Try larger speeds.
```

So:

```text
Yes → move right boundary left
No  → move left boundary right
```


# 11. Why a Working Speed Keeps the Candidate

Suppose speed 5 works.

We cannot throw away 5 because it might be the minimum answer.

Therefore:

```text
right = speed
```

not:

```text
right = speed - 1
```

If speed 5 fails, then 5 cannot be the answer, so we can safely move past it:

```text
left = speed + 1
```


# 12. Why the Search is Monotonic

If speed 5 works, then speed 6 also works.

A faster speed cannot require more hours than a slower speed.

So the answers look like:

```text
No No No No Yes Yes Yes Yes
                ↑
              answer
```

Binary Search is ideal for finding this first Yes.


# 13. Ceiling Division

For each pile, the required hours are:

```text
ceil(pile / speed)
```

For positive integers, we can calculate this without floating point:

```text
(pile + speed - 1) // speed
```

Example:

```text
pile = 11
speed = 3

(11 + 3 - 1) // 3
= 13 // 3
= 4
```

So the pile needs 4 hours.


# 14. Complete Solution Logic

The algorithm is:

```text
1. Set left = 1.
2. Set right = largest pile.
3. Pick the middle speed.
4. Calculate total required hours.
5. If hours <= h:
       speed works
       try smaller
6. Otherwise:
       speed is too slow
       try larger
7. Continue until left == right.
8. Return left.
```

The answer is the smallest feasible speed.


````python
class Solution:
    def minEatingSpeed(self, piles, h):
        left = 1
        right = max(piles)

        while left < right:
            speed = left + (right - left) // 2

            hours = 0

            for pile in piles:
                hours += (pile + speed - 1) // speed

            if hours <= h:
                # Speed works, so try a smaller speed.
                right = speed
            else:
                # Speed is too slow, so increase it.
                left = speed + 1

        return left
````

# 15. Understanding the Code

```text
left = 1
```

Smallest possible speed.

```text
right = max(piles)
```

Largest useful speed.

```text
speed = left + (right - left) // 2
```

Middle candidate.

Then we calculate how many hours all piles need.

If:

```text
hours <= h
```

the speed works, so we search smaller.

If:

```text
hours > h
```

the speed is too slow, so we search larger.


# 16. Why `left < right`?

This version searches for the smallest feasible value.

When:

```text
left == right
```

only one possible answer remains.

Therefore the loop stops and:

```text
return left
```

returns the minimum feasible speed.

# 20. Complexity

Let:

```text
n = number of piles
M = largest pile
```

Binary Search examines about:

```text
O(log M)
```

candidate speeds.

For each candidate, we inspect all `n` piles:

```text
O(n)
```

Therefore:

```text
Time Complexity = O(n log M)
Space Complexity = O(1)
```

This is much faster than checking every possible speed.


# 21. Normal Binary Search vs Binary Search on Answer

### Normal Binary Search

```text
Sorted array
    ↓
Find target
```

### Binary Search on Answer

```text
Possible answer range
    ↓
Check candidate
    ↓
Yes / No
    ↓
Find boundary
```

For Koko, we are searching the speed range, not searching through the pile values themselves.


# 22. The General Pattern

When a problem asks for a minimum or maximum, ask:

```text
1. What is the possible answer range?
2. Can I test one candidate answer?
3. Does the test return Yes/No?
4. Is that result monotonic?
5. Am I looking for the first Yes or last Yes?
```

For a minimum feasible answer:

```text
No No No Yes Yes Yes
        ↑
      answer
```

This pattern is the heart of Binary Search on Answer.


# 23. Common Mistakes

### Mistake 1: Searching the input array

For Koko, the search space is possible eating speeds.

### Mistake 2: Forgetting to round hours up

A partial final batch still needs a whole hour.

### Mistake 3: Throwing away a working speed

If a candidate works, it may still be the answer.

Keep it with:

```text
right = speed
```

### Mistake 4: Reversing the direction

Too many hours:

```text
increase speed
```

Few enough hours:

```text
try a smaller speed
```

### Mistake 5: Missing monotonicity

The method works because:

```text
No → No → No → Yes → Yes → Yes
```


# 24. Final Memory Trick

Remember:

```text
ANSWER RANGE
     ↓
PICK MIDDLE
     ↓
CHECK FEASIBILITY
     ↓
YES → TRY SMALLER
NO  → TRY LARGER
     ↓
REPEAT
```

For Koko:

```text
speed
  ↓
calculate required hours
  ↓
hours <= h ?
  ↓
Yes → speed works
No  → speed is too slow
```

The most important idea:

**We are not searching for the answer directly. We are searching for the boundary between impossible and possible answers.**

For LeetCode 875:

```text
Time:  O(n log M)
Space: O(1)
```

where `n` is the number of piles and `M` is the largest pile.
