# 1. What is a Trie?

A Trie is a special tree used mainly for words and strings.

Instead of storing a whole word as one value, a Trie stores the word character by character.

For example:

```text
cat
car
care
```

These words share the prefix:

```text
ca
```

A Trie reuses that shared path.

This is why it is called a **Prefix Tree**.


# 2. Prefix

A prefix is the beginning part of a word.

For:

```text
computer
```

some prefixes are:

```text
c
co
com
comp
compu
comput
compute
computer
```

For:

```text
cat
car
care
```

the shared prefix is:

```text
ca
```

A Trie stores shared prefixes only once.


# 3. Real-Life Example

Imagine a search box.

You type:

```text
ca
```

The system may suggest:

```text
cat
car
care
camera
cake
```

All of these words begin with `ca`.

A Trie is very useful for autocomplete and prefix-based search.


# 4. Trie Structure

For the word:

```text
cat
```

the path is:

```text
root
 |
 c
 |
 a
 |
 t
```

Now insert:

```text
car
```

The `c` and `a` nodes can be shared:

```text
      root
        |
        c
        |
        a
       / \
      t   r
```

This shared-prefix behavior is the key advantage of a Trie.


# 5. Trie Node

A Trie node usually stores:

```text
children
is_end
```

`children` stores the next possible characters.

`is_end` tells us whether a complete word ends at this node.

This second piece is important because one word can be a prefix of another.

For example:

```text
app
apple
```

The path for `app` is also part of the path for `apple`, but `app` may or may not be a complete inserted word.


# 6. Why `is_end` is Needed

Suppose the Trie contains only:

```text
apple
```

The path:

```text
a → p → p
```

exists.

But that does not mean:

```text
app
```

was inserted as a complete word.

So:

```text
search("app") → False
startsWith("app") → True
```

This gives us the important distinction:

```text
search      = exact word
startsWith  = prefix
```


# 7. The Three Main Operations

A Trie commonly supports:

## Insert

Add a complete word.

## Search

Check whether a complete word exists.

## startsWith

Check whether any inserted word begins with a prefix.

Example, after inserting:

```text
apple
application
```

we have:

```text
search("apple") → True
search("app") → False   # if app was not separately inserted
startsWith("app") → True
```


# 8. Insert Operation

To insert a word:

```text
Start at root

For each character:
    If the child does not exist:
        create it

    Move to that child

After the last character:
    mark the node as an end of a word
```

For:

```text
cat
```

we create or reuse:

```text
c → a → t
```

Then mark `t` as:

```text
is_end = True
```


# 9. Insert Multiple Words

Insert:

```text
cat
car
care
```

The Trie shares:

```text
c → a
```

A simplified picture:

```text
root
 |
 c
 |
 a
 / \
t   r
    |
    e
```

The end markers tell us which paths represent complete words.

```text
cat  → end at t
car  → end at r
care → end at e
```


# 10. Search Operation

To search for a complete word:

```text
Start at root

For each character:
    If it is missing:
        return False

    Move to the child

After all characters:
    return is_end
```

The final `is_end` check is what separates an exact word from a prefix.


# 11. `startsWith` Operation

`startsWith` asks:

> Does at least one stored word begin with this prefix?

For inserted words:

```text
apple
application
banana
```

we get:

```text
startsWith("app")  → True
startsWith("appl") → True
startsWith("ban")  → True
startsWith("cat")  → False
```

For `startsWith`, we only need the prefix path to exist.

We do not need the final node to be marked `is_end`.

# 14. Complexity

Let:

```text
L = length of the word or prefix
```

Then:

```text
insert      → O(L)
search      → O(L)
startsWith  → O(L)
```

Why?

Because each operation processes at most one Trie node per character.

If `N` is the total number of characters inserted across all words, the worst-case space is:

```text
O(N)
```

Shared prefixes can reduce the actual number of created nodes.


# 15. Trie vs Hash Set

A hash set is excellent for exact lookup:

```text
Does "apple" exist?
```

A Trie is especially useful when we also need prefix lookup:

```text
Does any word start with "app"?
```

So:

```text
Exact word lookup → Hash Set can be excellent
Prefix lookup     → Trie is especially useful
```

A Trie can support both exact word lookup and prefix queries.

# 17. Common Mistakes

### Mistake 1
Forgetting `is_end`.

A path existing does not necessarily mean the complete word exists.

### Mistake 2
Confusing:

```text
search
```

with:

```text
startsWith
```

### Mistake 3
Creating duplicate nodes for an existing prefix.

Reuse existing nodes.

### Mistake 4
Moving to a child without checking whether it exists.

### Mistake 5
Forgetting that one word can be a prefix of another.

Example:

```text
app
apple
```


# 18. Final Memory Trick

Think of a Trie as:

```text
WORD
 ↓
CHARACTERS
 ↓
SHARED PREFIX
 ↓
TREE PATH
 ↓
END MARKER
```

For LeetCode 208:

```text
insert(word)
    ↓
create/reuse path
    ↓
mark final node
```

```text
search(word)
    ↓
follow complete path
    ↓
check end marker
```

```text
startsWith(prefix)
    ↓
follow prefix path
    ↓
path exists = True
```

The most important idea:

**A prefix path existing does not automatically mean that the prefix is a complete word.**

# 19. Complete LeetCode 208 Solution

Use the following class directly on LeetCode.


````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False


class Trie:

    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        current = self.root

        for ch in word:
            if ch not in current.children:
                current.children[ch] = TrieNode()

            current = current.children[ch]

        current.is_end = True

    def search(self, word):
        current = self.root

        for ch in word:
            if ch not in current.children:
                return False

            current = current.children[ch]

        return current.is_end

    def startsWith(self, prefix):
        current = self.root

        for ch in prefix:
            if ch not in current.children:
                return False

            current = current.children[ch]

        return True
````
