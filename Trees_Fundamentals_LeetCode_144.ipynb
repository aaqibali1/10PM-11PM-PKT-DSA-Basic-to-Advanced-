# Trees: Fundamentals & Traversals

## LeetCode 144 — Binary Tree Preorder Traversal

Trees are a hierarchical data structure made of connected nodes. This notebook covers root, parent, child, leaf, depth, height, binary trees, preorder/inorder/postorder traversals, recursion, iterative traversal, and LeetCode 144.

Problem: https://leetcode.com/problems/binary-tree-preorder-traversal/


# 1. What is a Tree?

A tree is a collection of connected nodes arranged in a hierarchy.

```text
        A
       / \
      B   C
     / \
    D   E
```

Think about a family tree or an organization's management structure. One node can branch into smaller parts.


# 2. Root, Parent, Child, Sibling, Leaf

For:

```text
        A
       / \
      B   C
     / \
    D   E
```

**Root:** `A`, the top node.

**Parent:** `A` is parent of `B` and `C`; `B` is parent of `D` and `E`.

**Child:** `B` and `C` are children of `A`.

**Siblings:** `B` and `C` share the same parent. `D` and `E` also share the same parent.

**Leaf:** A node with no children. Here `C`, `D`, and `E` are leaves.


# 3. Edge, Depth, and Height

An **edge** is the connection between two nodes.

Depth tells how far a node is from the root. With the common edge-based definition:

```text
        A        depth 0
       / \
      B   C      depth 1
     / \
    D   E        depth 2
```

Height measures the longest path from a node down to a leaf. For example, `A → B → D` has 2 edges, so the tree height is 2 under the edge-based convention.


# 4. Binary Tree

A binary tree is a tree where each node has at most two children: a left child and a right child.

```text
        1
       / \
      2   3
     / \
    4   5
```

A node may have zero, one, or two children.


# 5. What is Tree Traversal?

Traversal means visiting every node in a specific order. The three depth-first traversals are:

```text
Preorder  = Root → Left → Right
Inorder   = Left → Root → Right
Postorder = Left → Right → Root
```


# 6. Preorder Traversal

Preorder visits the current node first:

```text
Root → Left → Right
```

For:

```text
          1
        /   \
       2     3
      / \\   / \
     4   5 6   7
```

The preorder result is:

```text
1, 2, 4, 5, 3, 6, 7
```

Memory trick: **PRE = root first**.


# 7. Inorder Traversal

Inorder visits the root between the two subtrees:

```text
Left → Root → Right
```

For the same tree:

```text
4, 2, 5, 1, 6, 3, 7
```

In a binary search tree, inorder traversal produces values in sorted order.


# 8. Postorder Traversal

Postorder visits the root last:

```text
Left → Right → Root
```

For the same tree:

```text
4, 5, 2, 6, 7, 3, 1
```


# 9. Easy Traversal Memory Trick

For:

```text
          Root
         /    \
       Left   Right
```

Remember where the root appears:

```text
Preorder  → Root first
Inorder   → Root in the middle
Postorder → Root last
```


# 10. Trees and Recursion

Trees are naturally recursive because each child can be treated as the root of a smaller tree.

For preorder, the rule is always:

```text
1. Visit current node
2. Traverse left subtree
3. Traverse right subtree
```

If the current node is `None`, there is nothing to visit, so we stop.


# 11. LeetCode 144 — Binary Tree Preorder Traversal

The problem gives the root of a binary tree and asks for the node values in preorder.

Example:

```text
      1
       \
        2
       /
      3
```

Preorder:

```text
1, 2, 3
```

The main rule remains:

```text
Root → Left → Right
```


````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def preorderTraversal(self, root):
        result = []

        def preorder(node):
            if node is None:
                return

            result.append(node.val)
            preorder(node.left)
            preorder(node.right)

        preorder(root)
        return result
````

# 12. Understanding the Recursive Solution

`result` stores the answer.

For each node:

```text
result.append(node.val) → visit root first
preorder(node.left)      → process left
preorder(node.right)     → process right
```

This is exactly the definition of preorder traversal.

# 14. Iterative Preorder Traversal

We can also use an explicit stack instead of recursion.

A stack is LIFO: last in, first out. To process the left child before the right child, push the right child first and the left child second.

For:

```text
    1
   / \
  2   3
```

After visiting `1`, push `3` then `2`. The stack removes `2` first, giving the preorder `1, 2, 3`.


````python
class Solution:
    def preorderTraversal(self, root):
        if root is None:
            return []

        result = []
        stack = [root]

        while stack:
            node = stack.pop()
            result.append(node.val)

            if node.right is not None:
                stack.append(node.right)
            if node.left is not None:
                stack.append(node.left)

        return result
````

# 15. Build a Tree in Google Colab

Example tree:

```text
          1
        /   \
       2     3
      / \\   / \
     4   5 6   7
```


````python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.left = TreeNode(6)
root.right.right = TreeNode(7)

solution = Solution()
print(solution.preorderTraversal(root))
````

# 16. Expected Output

```text
[1, 2, 4, 5, 3, 6, 7]
```


````python
# More test cases

solution = Solution()

# Empty tree
print("Empty:", solution.preorderTraversal(None))

# One node
one = TreeNode(10)
print("One node:", solution.preorderTraversal(one))

# Only left children
left_tree = TreeNode(1)
left_tree.left = TreeNode(2)
left_tree.left.left = TreeNode(3)
print("Only left:", solution.preorderTraversal(left_tree))

# Only right children
right_tree = TreeNode(1)
right_tree.right = TreeNode(2)
right_tree.right.right = TreeNode(3)
print("Only right:", solution.preorderTraversal(right_tree))
````

# 17. Complexity

For a tree with `n` nodes, every node is visited exactly once.

```text
Time Complexity = O(n)
```

The recursive solution uses call-stack space. In a completely skewed tree, recursion depth can reach `n`, so the worst-case auxiliary space is:

```text
O(n)
```

For a balanced tree, the recursion depth is about `O(log n)`.


# 18. Common Mistakes

1. Mixing up the traversal orders.

```text
Preorder  = Root Left Right
Inorder   = Left Root Right
Postorder = Left Right Root
```

2. Forgetting the `None` base case.

3. Visiting the root too late. Preorder means the root comes first.

4. In iterative preorder, pushing left before right. Because the stack is LIFO, push right first.

5. Confusing height measured in edges with height measured in nodes.


# 19. Final Cheat Sheet

Tree terms:

```text
Root   → top node
Parent → node above
Child  → node below
Leaf   → node with no children
Depth  → distance from root
Height → longest downward path
```

Traversals:

```text
Preorder  = Root → Left → Right
Inorder   = Left → Root → Right
Postorder = Left → Right → Root
```

LeetCode 144:

```text
Visit current node
→ traverse left subtree
→ traverse right subtree
```

Time complexity:

```text
O(n)
```

The most important idea: **Preorder means visit the current node first, then completely process its left subtree, then its right subtree.**
