## 1. What is a Linked List?

A **linked list** is a sequence of **nodes**. Each node stores:

1. **Data** — the actual value.
2. **Next** — a reference to the next node.

Unlike an array/list, linked-list nodes do not need to be next to each other in memory. They are connected by references.

### Simple picture

```text
head
 ↓
[10 | next] → [20 | next] → [30 | None]
```

- `head` points to the first node.
- The last node points to `None`, which means the list has ended.


### Real-life example: a treasure hunt

Imagine a treasure hunt. At each location you find:

- some information, and
- a clue telling you where to go next.

That is similar to a linked list:

```text
Location 1 → Location 2 → Location 3 → End
```

A node is like a location, and `next` is like the clue to the next location.

## 2. Create a Node in Python

In Python, we can make a node using a class.

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


# Create three nodes
node1 = ListNode(10)
node2 = ListNode(20)
node3 = ListNode(30)

# Connect them
node1.next = node2
node2.next = node3

# The head is the first node
head = node1

print(head.val)
print(head.next.val)
print(head.next.next.val)
````

## 3. Traversal — Visiting Every Node

**Traversal** means starting from the head and visiting nodes one by one until `None`.

### Important rule
Do not move the real `head` while traversing. Use a temporary variable such as `current` or `temp`.

Real-life idea: if the first page of a notebook is your starting point, you should use a bookmark while reading instead of moving the notebook's original first-page reference.

````python
def traverse_list(head):
    current = head
    while current is not None:
        print(current.val, end=" → ")
        current = current.next
    print("None")


traverse_list(head)
````

**Time complexity:** `O(n)` because, in the worst case, we visit all `n` nodes.

## 4. Insertion at the Beginning

Suppose we have:

```text
10 → 20 → 30 → None
```

We want to insert `5` at the beginning.

Only two links matter:

```text
5.next = head
head = 5
```

Result:

```text
5 → 10 → 20 → 30 → None
```

### Real-life example
Imagine a line of people and a new person joins at the **front**. You do not need to move everyone; you only change who is first and connect the new person to the old first person.

**Time complexity:** `O(1)`.

````python
def insert_at_head(head, value):
    new_node = ListNode(value)
    new_node.next = head
    return new_node


head = insert_at_head(head, 5)
traverse_list(head)
````

## 5. Insertion at the End

To insert at the end, we first walk to the last node and then connect the new node.

```text
10 → 20 → 30 → None
                  ↑
             add 40 here
```

### Real-life example
Like adding a person to the **end of a queue**: you have to reach the end first unless you already keep a direct reference to the tail.

**Time complexity:** `O(n)` without a tail pointer; `O(1)` when a tail pointer is maintained.

````python
def insert_at_end(head, value):
    new_node = ListNode(value)

    if head is None:
        return new_node

    current = head
    while current.next is not None:
        current = current.next

    current.next = new_node
    return head


head = insert_at_end(head, 40)
traverse_list(head)
````

## 6. Insertion at a Specific Position

For a simple 0-based position, to insert into the middle, we go to the node just **before** the insertion point.

Example:

```text
10 → 20 → 30
```

Insert `25` at position 2:

```text
10 → 20 → 25 → 30
```

The key pointer changes are:

```text
new_node.next = previous.next
previous.next = new_node
```

### Real-life example
Think of adding a person **between two people in a chain**. You tell the new person who comes next, then tell the previous person to point to the new person.

**Time complexity:** `O(n)` because we may need to walk to the required position.

````python
def insert_at_position(head, value, position):
    if position < 0:
        raise ValueError("Position cannot be negative")

    if position == 0:
        return insert_at_head(head, value)

    current = head
    for _ in range(position - 1):
        if current is None:
            raise IndexError("Position is out of range")
        current = current.next

    if current is None:
        raise IndexError("Position is out of range")

    new_node = ListNode(value)
    new_node.next = current.next
    current.next = new_node
    return head


head = insert_at_position(head, 25, 3)
traverse_list(head)
````

## 7. Deletion from the Beginning

To remove the first node, simply move `head` to the second node:

```python
head = head.next
```

### Real-life example
Like removing the first person from a queue: the second person becomes the new first person.

**Time complexity:** `O(1)`.

````python
def delete_from_head(head):
    if head is None:
        return None
    return head.next


head = delete_from_head(head)
traverse_list(head)
````

## 8. Deletion from the End

For a singly linked list, we normally need to find the **second-last node** and set its `next` to `None`.

### Real-life example
Imagine a chain of people holding hands. To remove the last person, you usually need to reach the person just before them and make that person's hand point to `nothing`.

**Time complexity:** `O(n)`.

````python
def delete_from_end(head):
    if head is None:
        return None
    if head.next is None:
        return None

    current = head
    while current.next.next is not None:
        current = current.next

    current.next = None
    return head


head = delete_from_end(head)
traverse_list(head)
````

## 9. Deletion at a Specific Position

To delete a node in the middle, we usually reach the node **before** it and skip the target.

The important idea is:

```python
previous.next = previous.next.next
```

Example:

```text
10 → 20 → 30 → 40
```

Delete `30`:

```text
10 → 20 ─────────→ 40
```

### Real-life example
Like removing one person from a chain: the person before them connects directly to the person after them.

**Time complexity:** `O(n)` because we may need to find the target position.

````python
def delete_at_position(head, position):
    if position < 0:
        raise ValueError("Position cannot be negative")
    if head is None:
        return None
    if position == 0:
        return head.next

    current = head
    for _ in range(position - 1):
        if current.next is None:
            raise IndexError("Position is out of range")
        current = current.next

    if current.next is None:
        raise IndexError("Position is out of range")

    current.next = current.next.next
    return head


head = delete_at_position(head, 2)
traverse_list(head)
````

# 10. Important Problem: Reverse a Linked List

This problem asks us to change:

```text
1 → 2 → 3 → 4 → None
```

into:

```text
4 → 3 → 2 → 1 → None
```

The values do not need to move to new nodes. We simply **reverse the `next` links**.

### Real-life example: a chain of people

Imagine four people standing in a line, where each person points to the next person:

```text
1 → 2 → 3 → 4 → End
```

To reverse the chain, person 1 should point backward to nobody, person 2 should point to person 1, person 3 to person 2, and person 4 to person 3.

That is exactly what the algorithm does.

## 11. The Three Main Variables

We use:

```python
current = head
prev = None
```

- `current` = the node we are currently working on.
- `prev` = the node that should come before `current` after reversal.
- `next_node` = temporarily saves the original next node so we do not lose the rest of the list.

### The four steps to remember

```text
1. Save     → next_node = current.next
2. Reverse  → current.next = prev
3. Move prev→ prev = current
4. Move curr→ current = next_node
```


````python
def reverse_list(head):
    current = head
    prev = None

    while current:
        next_node = current.next   # 1. Save the next node
        current.next = prev        # 2. Reverse the link
        prev = current             # 3. Move prev forward
        current = next_node       # 4. Move current forward

    return prev


# Example
head = None
for value in [1, 2, 3, 4]:
    head = insert_at_end(head, value)

print("Before:")
traverse_list(head)

head = reverse_list(head)

print("After:")
traverse_list(head)
````

## 12. Reverse List — Easy Step-by-Step Trace

Start:

```text
prev = None
current = 1

1 → 2 → 3 → 4 → None
```

After processing `1`:

```text
1 → None
```

After processing `2`:

```text
2 → 1 → None
```

After processing `3`:

```text
3 → 2 → 1 → None
```

After processing `4`:

```text
4 → 3 → 2 → 1 → None
```

At the end, `current` becomes `None`, and `prev` is the new head (`4`). So we return `prev`.

## 13. Why Do We Save `next_node` First?

This line is critical:

```python
next_node = current.next
```

Suppose we have:

```text
1 → 2 → 3 → 4
```

If we immediately change `1.next` to `None` without first saving `2`, we no longer have a reference from node `1` to node `2`.

So the safe order is:

```text
Save the original next node
        ↓
Reverse the current link
        ↓
Move to the saved node
```

### Easy memory trick
**Save → Reverse → Move → Move**

## 14. Reverse List vs. Other Operations

| Operation | Main idea | Time |
|---|---|---:|
| Traversal | Visit nodes one by one | `O(n)` |
| Insert at head | New node becomes first | `O(1)` |
| Insert at end | Find last node, then link | `O(n)`* |
| Delete at head | Move head to next node | `O(1)` |
| Delete in middle | Skip the target node | `O(n)` |
| Reverse list | Turn every `next` link backward | `O(n)` |

`*` With a maintained tail pointer, insertion at the end can be `O(1)`.

### The key difference

Most linked-list operations are about **changing links** rather than shifting all elements like an array might require.

# 15. Quick Revision

### Remember these words

- **Node** → stores data + reference to next node
- **Head** → first node
- **Tail** → last node (conceptually; a list may or may not store a direct tail pointer)
- **None** → end of the list
- **Traversal** → visit nodes from head to end
- **Insertion** → connect a new node into the chain
- **Deletion** → bypass a node in the chain
- **Reverse** → change every link to point backward

### Best mental model
Think of a linked list as a **chain of people holding hands**. Each person knows who comes next. Most operations simply change who is connected to whom.
