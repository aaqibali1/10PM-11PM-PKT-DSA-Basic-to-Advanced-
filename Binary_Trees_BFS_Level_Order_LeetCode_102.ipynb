# 1. What is BFS?

BFS means:

**Breadth-First Search**

It explores a tree **level by level**.

For example:

```text
          1
        /   \
       2     3
      / \   / \
     4   5 6   7
```

BFS visits:

```text
1
2, 3
4, 5, 6, 7
```

So the order is:

```text
1, 2, 3, 4, 5, 6, 7
```


# 2. What Does Level Order Mean?

Level order means:

> Visit all nodes on one level before moving to the next level.

For this tree:

```text
          1          <- Level 0
        /   \
       2     3       <- Level 1
      / \   / \
     4   5 6   7     <- Level 2
```

The levels are:

```text
Level 0 -> [1]
Level 1 -> [2, 3]
Level 2 -> [4, 5, 6, 7]
```

The answer is therefore:

```text
[[1], [2, 3], [4, 5, 6, 7]]
```


# 3. BFS vs DFS

You already learned DFS traversals:

```text
Preorder
Inorder
Postorder
```

DFS goes deep into a branch before moving to another branch.

BFS works differently.

BFS goes:

```text
Across the current level
        ↓
Then the next level
        ↓
Then the next level
```

Think:

```text
DFS -> go deep
BFS -> go wide
```


# 4. Real-Life Example

Imagine an organization:

```text
CEO
├── Manager A
│   ├── Employee 1
│   └── Employee 2
└── Manager B
    ├── Employee 3
    └── Employee 4
```

Level order would be:

```text
CEO

Manager A, Manager B

Employee 1, Employee 2, Employee 3, Employee 4
```

You finish one management level before moving to the next.

That is BFS.


# 5. Why Does BFS Use a Queue?

A queue follows:

```text
FIFO
First In, First Out
```

Suppose we start with:

```text
[1]
```

We remove `1` and then add its children:

```text
[2, 3]
```

We remove `2` next, then add its children:

```text
[3, 4, 5]
```

We remove `3` next, then add its children:

```text
[4, 5, 6, 7]
```

The queue naturally keeps nodes in level order.

# 7. Why Queue Gives Level Order

Notice the pattern:

```text
Parent enters queue
        ↓
Parent is removed
        ↓
Children enter queue
        ↓
Children are processed after earlier nodes
```

Because the queue is FIFO, older nodes leave first.

That naturally causes nodes to be processed from left to right, level by level.


# 8. Level Processing

LeetCode 102 wants separate groups for each level.

Example:

```text
        1
       / \
      2   3
     / \   \
    4   5   6
```

Output:

```text
[
    [1],
    [2, 3],
    [4, 5, 6]
]
```

So we need to know how many nodes belong to the current level.


# 9. How Do We Know the Size of a Level?

Suppose the queue currently contains:

```text
[2, 3]
```

Both nodes belong to the same level.

So before processing them, we record:

```text
level_size = 2
```

Then we process exactly 2 nodes.

While processing those nodes, their children are added to the queue.

Those children belong to the next level.

This is the key level-order technique.

# 11. The Main BFS Pattern

Remember this:

```text
Put root in queue
      ↓
While queue is not empty
      ↓
Count current level size
      ↓
Process exactly that many nodes
      ↓
Add their children
      ↓
Save the level
```

This is one of the most important Binary Tree BFS patterns.

# 15. Why We Process Exactly the Current Level Size

Suppose:

```text
Queue = [9, 20]
```

Then:

```text
level_size = 2
```

We process exactly two nodes.

During that process, `15` and `7` may be added.

The queue becomes:

```text
[15, 7]
```

But these nodes belong to the next level.

That is why we must save the original queue size before processing the level.


# 16. Python Queue for BFS

In Python, `deque` is a good queue implementation.

Important operations:

```text
append()  -> add to the back
popleft() -> remove from the front
```

These match FIFO queue behavior.


````python
from collections import deque
````

````python
# LeetCode 102 - Binary Tree Level Order Traversal

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def levelOrder(self, root):
        if root is None:
            return []

        result = []
        queue = deque([root])

        while queue:
            level = []
            level_size = len(queue)

            for _ in range(level_size):
                node = queue.popleft()
                level.append(node.val)

                if node.left is not None:
                    queue.append(node.left)

                if node.right is not None:
                    queue.append(node.right)

            result.append(level)

        return result
````

# 17. Understanding the Code

The queue starts with:

```text
[root]
```

Then:

```text
level_size = len(queue)
```

tells us how many nodes belong to the current level.

For each node in that level:

```text
popleft()
```

gets the next node.

Then:

```text
level.append(node.val)
```

stores its value.

Its children are added to the queue for the next level.

Finally:

```text
result.append(level)
```

stores the completed level.

# 22. BFS vs DFS

## BFS

```text
Level by level
```

Uses:

```text
Queue
```

```text
Level order of trees

```

## DFS

```text
Go deep first
```

```text
Preorder

```

DFS commonly uses recursion or a stack.


# 23. Complexity

Let `n` be the number of nodes.

Every node enters the queue once and leaves the queue once.

Therefore:

```text
Time Complexity = O(n)
```

The queue can hold many nodes at the same time.

Worst-case auxiliary space:

```text
O(n)
```

So:

```text
Time  = O(n)
Space = O(n)
```


# 24. Common Mistakes

### Mistake 1
Using a stack instead of a queue.

BFS needs FIFO behavior.

### Mistake 2
Forgetting the current level size.

Without:

```text
level_size = len(queue)
```

different levels can be mixed together.

### Mistake 3
Processing newly added children immediately.

Children added during the current level belong to the next level.

### Mistake 4
Forgetting the empty tree.

If root is `None`, return:

```text
[]
```

### Mistake 5
Mixing BFS with preorder.

Preorder is:

```text
Root → Left → Right
```

Level order is:

```text
Level 0 → Level 1 → Level 2 → ...
```


# 25. Final Memory Trick

Remember Level Order as:

```text
ROOT
  ↓
QUEUE
  ↓
CURRENT LEVEL SIZE
  ↓
PROCESS THAT MANY NODES
  ↓
ADD THEIR CHILDREN
  ↓
NEXT LEVEL
```

For LeetCode 102:

```text
BFS = Queue + len(queue) + level list
```

The key idea:

**Process exactly the nodes that were already in the queue at the beginning of the level. Any children added during that process belong to the next level.**

Complexity:

```text
Time  = O(n)
Space = O(n)
```
