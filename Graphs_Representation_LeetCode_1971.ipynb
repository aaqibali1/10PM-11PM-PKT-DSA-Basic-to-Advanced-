# 1. What is a Graph?

A graph is a data structure used to represent relationships.

It has:

```text
Vertices (nodes)
+
Edges (connections)
```

Example:

```text
A ----- B
|       |
|       |
C ----- D
```

The letters are vertices.

The lines are edges.

Graphs can represent cities and roads, computers and connections, people and friendships, and many other relationships.


# 2. Vertices and Edges

A **vertex** is an object in the graph.

An **edge** is a connection between two vertices.

Example:

```text
A ----- B
```

Here:

```text
A = vertex
B = vertex
A-B = edge
```

A graph is basically:

```text
objects + relationships
```


# 3. Directed Graph

A directed graph has direction.

Example:

```text
A → B
```

This means there is a connection from `A` to `B`.

It does not automatically mean there is a connection from `B` to `A`.

Examples in real life include following relationships and one-way roads.


# 4. Undirected Graph

An undirected graph has a two-way connection.

Example:

```text
A — B
```

This means:

```text
A → B
B → A
```

A two-way road is a good real-life example.

LeetCode 1971 uses an **undirected** graph.


# 5. Weighted vs Unweighted

An **unweighted** graph only tells us that a connection exists:

```text
A — B
```

A **weighted** graph also stores a number such as distance, cost, or time:

```text
A --5-- B
```

Here the edge weight is `5`.

LeetCode 1971 uses an **unweighted** graph.


# 6. Adjacency List

An adjacency list stores the neighbors of every vertex.

Example:

```text
A ----- B
|       |
C ----- D
```

Adjacency list:

```text
A → [B, C]
B → [A, D]
C → [A, D]
D → [B, C]
```

This is one of the most common graph representations in DSA.


````python
# Adjacency list with a dictionary

graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

print("Neighbors of A:", graph["A"])
````

# 7. Building an Undirected Adjacency List

For an undirected edge:

```text
A — B
```

we store both directions:

```text
A → B
B → A
```

So for:

```text
A-B
A-C
B-D
C-D
```

we get:

```text
A → [B, C]
B → [A, D]
C → [A, D]
D → [B, C]
```


````python
edges = [
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("C", "D")
]

graph = {node: [] for node in ["A", "B", "C", "D"]}

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)

print(graph)
````

# 8. Adjacency Matrix

An adjacency matrix uses a table.

For:

```text
A — B
|
C
```

we can use:

```text
    A B C
A   0 1 1
B   1 0 0
C   1 0 0
```

`1` means an edge exists.

`0` means there is no edge.


# 9. Directed Adjacency Matrix

For:

```text
A → B
B → C
```

the matrix can be:

```text
    A B C
A   0 1 0
B   0 0 1
C   0 0 0
```

The matrix is not necessarily symmetric because direction matters.


# 10. Weighted Adjacency Matrix

For a weighted graph:

```text
A --5-- B
A --2-- C
```

the matrix can store the weights:

```text
    A B C
A   0 5 2
B   5 0 0
C   2 0 0
```

For LeetCode 1971, weights are unnecessary.


# 11. Adjacency List vs Matrix

### Adjacency List

Stores actual connections.

Typical space:

```text
O(V + E)
```

Good for sparse graphs and BFS/DFS.

### Adjacency Matrix

Stores every possible pair.

Space:

```text
O(V²)
```

Useful when we frequently need to check whether a direct edge exists.


# 12. BFS

BFS means Breadth-First Search.

It explores outward using a queue:

```text
start
 ↓
neighbors
 ↓
neighbors of neighbors
 ↓
...
```

BFS processes nodes by increasing distance from the start in an unweighted graph.

It is useful for path and shortest-distance problems.


# 13. DFS

DFS means Depth-First Search.

It explores one path deeply before coming back.

It can use:

```text
recursion
```

or:

```text
stack
```

Both BFS and DFS can answer the basic question:

> Can I reach the destination?


# 14. Why `visited` is Needed

Graphs can contain cycles.

Example:

```text
0 ----- 1
 \     /
   2
```

From `0`, we might visit:

```text
0 → 1 → 2 → 0 → ...
```

Without a visited set, we could revisit the same nodes forever.

So when a node is discovered, mark it visited.

# 20. BFS Solution Pattern

The standard pattern is:

```text
Build adjacency list
Create queue
Mark source visited

While queue is not empty:
    remove front node

    if it is destination:
        return True

    for each neighbor:
        if not visited:
            mark visited
            add to queue

return False
```


````python
from collections import deque


class Solution:
    def validPath(self, n, edges, source, destination):
        # Build adjacency list
        graph = [[] for _ in range(n)]

        for u, v in edges:
            graph[u].append(v)
            graph[v].append(u)

        # BFS
        queue = deque([source])
        visited = {source}

        while queue:
            current = queue.popleft()

            if current == destination:
                return True

            for neighbor in graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)

        return False
````

# 21. Why Mark a Node Visited When Adding It?

Suppose two nodes both connect to the same neighbor.

If we wait until removing that neighbor from the queue, it could be added multiple times.

Instead, as soon as we discover it:

```text
mark visited
add to queue
```

Then it can only be added once.

This improves efficiency and prevents repeated exploration.


# 22. DFS Solution

The same problem can be solved with DFS.

Conceptually:

```text
Start at source
↓
Mark current visited
↓
Explore every unvisited neighbor
↓
If destination is found → True
```

For connectivity, BFS and DFS are both valid.


````python
class Solution:
    def validPath(self, n, edges, source, destination):
        graph = [[] for _ in range(n)]

        for u, v in edges:
            graph[u].append(v)
            graph[v].append(u)

        visited = set()

        def dfs(node):
            if node == destination:
                return True

            visited.add(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    if dfs(neighbor):
                        return True

            return False

        return dfs(source)
````

For LeetCode 1971, either works.

# 27. Complexity

Let:

```text
V = number of vertices
E = number of edges
```

Building the adjacency list:

```text
O(V + E)
```

BFS or DFS:

```text
O(V + E)
```

because every vertex is visited at most once and every edge is examined a limited number of times.

Space:

```text
O(V + E)
```

for the adjacency list, plus the visited structure and traversal storage.

This is the standard complexity for graph traversal using an adjacency list.

# 31. Common Mistakes

### Mistake 1
Forgetting both directions in an undirected graph.

### Mistake 2
Not using `visited` in graphs with cycles.

### Mistake 3
Using a normal list with `pop(0)` for large BFS workloads.

### Mistake 4
Confusing directed and undirected edges.

### Mistake 5
Using an adjacency matrix when an adjacency list is more appropriate.

### Mistake 6
Forgetting that `source == destination` is already a valid path.
