# Binary Search

## LeetCode 704 — Binary Search

Binary Search is a fast searching algorithm that repeatedly cuts a sorted search space roughly in half.

Main ideas:
- Sorted data
- Search space
- Middle element
- Eliminating half of the search space
- Invariants
- Time complexity: O(log n)

Problem:
https://leetcode.com/problems/binary-search/

## 1. What is Searching?

Searching means trying to find a particular item in a collection.

Example:

```text
[10, 20, 30, 40, 50]
```

We want to find `40`. The goal is to determine whether it exists and, in LeetCode 704, return its position.

## 2. Linear Search

A simple method checks values one by one.

```text
10 → not 40
20 → not 40
30 → not 40
40 → found
```

This is Linear Search and takes O(n) time.

## 3. The Big Idea Behind Binary Search

Binary Search uses sorted order.

Example:

```text
[10, 20, 30, 40, 50, 60, 70]
```

Find `60`. The middle is `40`. Since `60 > 40`, everything on the left can be ignored. Search only:

```text
[50, 60, 70]
```

One comparison removed about half the search space.

## 4. Real-Life Example

Imagine finding a name in an alphabetically sorted dictionary. If you open near `M` while searching for `Zara`, you know the target must be later. You can ignore the first half. Repeating this is Binary Search.

## 5. Binary Search Requires Sorted Data

Binary Search depends on ordering.

Sorted:
```text
[10, 20, 30, 40, 50, 60]
```

Unsorted:
```text
[30, 10, 50, 20, 60, 40]
```

In the unsorted version, comparing the middle value does not tell us which half can be safely discarded.

## 6. Search Space

The search space is the part where the answer could still be.

Initially the whole array is possible. After checking the middle, we remove the half where the target cannot be.

The goal is to keep shrinking the search space until we find the target or nothing remains.

## 7. The Middle Element

At each step, inspect the middle of the current search space. There are three cases:

```text
target == middle → found
target < middle  → search left
target > middle  → search right
```

## 8. Step-by-Step Example

Find `60` in:

```text
[10, 20, 30, 40, 50, 60, 70]
```

Middle is `40`. Since `60 > 40`, search the right half:

```text
[50, 60, 70]
```

Middle is now `60`, so it is found at index `5`.

## 9. Another Example

Find `20` in:

```text
[10, 20, 30, 40, 50, 60, 70]
```

Middle is `40`. Since `20 < 40`, search the left half:

```text
[10, 20, 30]
```

Middle is `20`, so it is found at index `1`.

## 10. When the Target Does Not Exist

Find `35` in:

```text
[10, 20, 30, 40, 50]
```

Check `30`: target is larger, so search right.

Check `40`: target is smaller, so search left.

Nothing remains, so return `-1`.

## 11. Two Boundaries

A standard implementation tracks:

```text
left
right
```

For `[10, 20, 30, 40, 50]`:

```text
left = 0
right = 4
```

Everything between `left` and `right` is the current search space.

## 12. Calculating the Middle

A common expression is:

```python
mid = left + (right - left) // 2
```

It finds the middle index without first adding the two boundaries.

## 13. Updating the Search Space

If `target > nums[mid]`, the middle and everything before it cannot be the answer, so:

```python
left = mid + 1
```

If `target < nums[mid]`:

```python
right = mid - 1
```

We remove positions that are proven impossible.

## 14. What is an Invariant?

An invariant is a condition that remains true while the algorithm runs.

For Binary Search:

> If the target exists, it is still inside the current search space between `left` and `right`.

This is the key correctness idea.

## 15. Why Binary Search is O(log n)

The search space is approximately halved each time:

```text
1,000,000
→ 500,000
→ 250,000
→ 125,000
→ ...
→ 1
```

The number of halvings grows like `log₂(n)`, so the time complexity is `O(log n)`.

## 16. LeetCode 704 — Binary Search

The problem gives a sorted array `nums` of distinct integers and a `target`. Return the target index, or `-1` if it does not exist.

Example:

```text
nums = [-1, 0, 3, 5, 9, 12]
target = 9
```

Answer:

```text
4
```

## 17. Full Walkthrough

For:

```text
nums = [-1, 0, 3, 5, 9, 12]
target = 9
```

Start `left = 0`, `right = 5`.

Middle index is `2`, value `3`. Since `9 > 3`, move right.

Now search indices `3..5`. Middle is index `4`, value `9`. Found.

Answer: `4`.

````python
class Solution:
    def search(self, nums, target):
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = left + (right - left) // 2

            if nums[mid] == target:
                return mid

            if nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return -1
````

## 18. Understanding the Code

`left` and `right` define the search space.

`mid` is the position we inspect.

If the middle value equals the target, return `mid`.

If the middle value is too small, move `left` rightward.

If the middle value is too large, move `right` leftward.

The loop continues while `left <= right`, because `left == right` means one possible position still needs to be checked.

## 19. Example Where Target is Missing

```text
nums = [1, 3, 5, 7, 9]
target = 6
```

Middle is `5`, so search right.

Middle of `[7, 9]` is `7`, so search left.

The search space becomes empty. Return `-1`.

## 21. Common Mistakes

1. Using Binary Search on unsorted data.
2. Forgetting `mid + 1` when moving right.
3. Forgetting `mid - 1` when moving left.
4. Using the wrong loop condition.
5. Losing the invariant: if the target exists, it must remain inside the current search space.

Remember that the middle has already been checked, so it can be excluded when moving left or right.

## 22. Search Space Thinking

A powerful DSA habit is to ask:

> What is my current search space?

Initially:
```text
Whole array
```

After a comparison:
```text
Left half
or
Right half
```

The algorithm works because impossible positions are continuously removed.

## 23. Binary Search Beyond LeetCode 704

The same idea appears in many harder problems:

- First or last occurrence
- Search in rotated sorted arrays
- Find a boundary
- Find a minimum or maximum that satisfies a condition
- Binary search on the answer

The bigger pattern is:

```text
Ordered search space
+
A way to eliminate half
=
Binary Search idea
```

## 24. Final Memory Trick

Remember Binary Search as:

```text
SORTED
   ↓
MIDDLE
   ↓
COMPARE
   ↓
CUT HALF
   ↓
REPEAT
```

Standard exact search:

```text
left = 0
right = n - 1

while left <= right:
    check middle

    equal → found
    target smaller → go left
    target larger  → go right
```

Most important complexity:

```text
Linear Search: O(n)
Binary Search: O(log n)
```

Most important correctness idea:

**If the target exists, never remove the part of the search space where it could still be.**
