# Binary Search Trees

## LeetCode 700 — Search in a Binary Search Tree

A Binary Search Tree (BST) is a binary tree with an ordering rule:

```text
Left subtree < Node < Right subtree
```

This rule makes search, insertion, and min/max operations efficient when the tree is reasonably balanced.

Problem:
https://leetcode.com/problems/search-in-a-binary-search-tree/


# 1. What is a Binary Search Tree?

A BST is a binary tree where values are arranged in a specific order.

Example:

```text
        8
       / \
      3   10
     / \    \
    1   6    14
```

For node `8`:

```text
left side  → smaller values
right side → larger values
```

The same rule continues inside every subtree.


# 2. The BST Invariant

An invariant is a rule that must remain true.

For a standard BST:

```text
Every value in the left subtree is smaller than the node.
Every value in the right subtree is larger than the node.
```

Example:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
```

This rule tells us where a target value can possibly be.


# 3. BST vs Normal Binary Tree

A normal binary tree only says each node has at most two children.

A BST adds an ordering rule.

Valid BST:

```text
    8
   / \
  3   10
```

Not a valid BST:

```text
    8
   / \
  3    5
```

The `5` is on the right of `8`, but it is smaller than `8`.

So the second structure is a binary tree, but not a valid BST.


# 4. Searching in a BST

At every node:

```text
target == current → found
target < current  → go left
target > current  → go right
```

Example:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
        / \
       4   7
```

Search for `7`:

```text
7 < 8 → left
7 > 3 → right
7 > 6 → right
7 = 7 → found
```

Path:

```text
8 → 3 → 6 → 7
```


# 5. Why BST Search Works

The BST invariant lets us throw away an entire subtree.

If:

```text
target < current
```

the target cannot be in the right subtree.

If:

```text
target > current
```

the target cannot be in the left subtree.

So every comparison tells us which direction to take.


# 6. Real-Life Example

Imagine a number guessing game.

You want to find a number between 1 and 100.

You guess 50.

If the answer is smaller, you ignore 51–100.

If the answer is larger, you ignore 1–49.

BST search works similarly:

```text
Compare
↓
Choose left or right
↓
Ignore the other side
```

# 8. Insert in a BST

Insertion uses the same left/right decisions.

Tree:

```text
        8
       / \
      3   10
     / \
    1   6
```

Insert `7`:

```text
7 < 8 → left
7 > 3 → right
7 > 6 → right
```

The right position of `6` is empty, so insert `7`.

Result:

```text
        8
       / \
      3   10
     / \
    1   6
         \
          7
```


# 9. Minimum and Maximum

In a BST:

```text
Minimum → farthest left
Maximum → farthest right
```

Example:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
```

Minimum:

```text
8 → 3 → 1
```

Maximum:

```text
8 → 10 → 14
```

So:

```text
min = 1
max = 14
```


# 10. BST and Inorder Traversal

A very important property:

> Inorder traversal of a valid BST produces values in sorted order.

For:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
```

Inorder:

```text
1, 3, 6, 8, 10, 14
```

So:

```text
Inorder of BST = sorted order
```


# 11. Balanced vs Unbalanced BST

Balanced:

```text
        8
      /   \
     4     12
    / \   /  \
   2   6 10  14
```

Very unbalanced:

```text
1
 \
  2
   \
    3
     \
      4
       \
        5
```

A balanced BST can have search around:

```text
O(log n)
```

A badly skewed BST can degrade to:

```text
O(n)
```


# 12. LeetCode 700 — Search in a Binary Search Tree

The problem gives:

```text
root
val
```

Find the node whose value equals `val`.

If found, return that node.

If not found, return:

```text
None
```

Example:

```text
      4
     / \
    2   7
   / \
  1   3
```

Searching for `2` returns the subtree rooted at `2`:

```text
    2
   / \
  1   3
```

The answer is the node/subtree, not just the number `2`.


# 13. Recursive Search Logic

For each current node:

```text
If node is None:
    return None

If node.val == target:
    return node

If target < node.val:
    search left

Otherwise:
    search right
```

This directly follows the BST invariant.


````
# LeetCode 700 - Recursive Solution

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def searchBST(self, root, val):
        if root is None:
            return None

        if root.val == val:
            return root

        if val < root.val:
            return self.searchBST(root.left, val)

        return self.searchBST(root.right, val)
````

# 14. Understanding the Recursive Solution

The algorithm is simply:

```text
Compare
↓
Choose direction
↓
Repeat on that subtree
```

For target `3`:

```text
4
↓
3 < 4 → left
↓
2
↓
3 > 2 → right
↓
3
↓
found
```


# 15. Iterative Search

The same search can be done with a loop.

Instead of recursively calling the function, we keep a `current` node and move it left or right.

This follows the same logic and does not use recursive calls.


````
# LeetCode 700 - Iterative Solution

class Solution:
    def searchBST(self, root, val):
        current = root

        while current is not None:
            if current.val == val:
                return current

            if val < current.val:
                current = current.left
            else:
                current = current.right

        return None
````

# 18. Build a BST in Google Colab

We will build:

```text
      4
     / \
    2   7
   / \
  1   3
```


````
root = TreeNode(4)

root.left = TreeNode(2)
root.right = TreeNode(7)

root.left.left = TreeNode(1)
root.left.right = TreeNode(3)
````

````
solution = Solution()

found = solution.searchBST(root, 2)

if found is not None:
    print("Found:", found.val)
    print("Left child:", found.left.val if found.left else None)
    print("Right child:", found.right.val if found.right else None)
else:
    print("Not found")
````

# 21. Complexity

Let `n` be the number of nodes.

For a balanced BST:

```text
Search → O(log n)
Insert → O(log n)
Min    → O(log n)
Max    → O(log n)
```

For a highly skewed BST:

```text
Search → O(n)
Insert → O(n)
Min    → O(n)
Max    → O(n)
```

These operations follow a path from the root, so their time depends mainly on the tree height.


# 22. Common Mistakes

### Mistake 1
Treating every binary tree as a BST.

A BST must maintain the ordering rule.

### Mistake 2
Choosing the wrong direction.

```text
target smaller → left
target larger  → right
```

### Mistake 3
Returning only the value.

LeetCode 700 expects the node/subtree.

### Mistake 4
Forgetting the empty case.

If the path reaches `None`, return `None`.

### Mistake 5
Assuming every BST search is O(log n).

That depends on the tree being balanced or suitably shaped.

# 24. Final Memory Trick

Think of a BST as a number-based decision tree:

```text
             current
            /       \
       smaller      larger
          ↓            ↓
        LEFT         RIGHT
```

For LeetCode 700:

```text
Start at root
     ↓
Compare target
     ↓
Found → return node
     ↓
Smaller → left
Larger  → right
     ↓
Repeat
```

The key idea:

**The BST invariant tells you which entire side can be ignored after every comparison.**

Balanced BST:

```text
O(log n)
```

Worst-case skewed BST:

```text
O(n)
```
