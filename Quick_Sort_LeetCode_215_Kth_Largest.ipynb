# Sorting: Quick Sort

## LeetCode 215 — Kth Largest Element in an Array

Quick Sort is based on **pivot + partition + divide and conquer**.

Main ideas:
- Choose a pivot
- Partition values around it
- Recursively solve the smaller parts
- Average time: O(n log n)
- Worst-case time: O(n²)

LeetCode 215 uses the same partition idea, but the most efficient target is usually **Quickselect**: instead of sorting both sides, it continues only into the side containing the answer.

Problem: https://leetcode.com/problems/kth-largest-element-in-an-array/

# 1. What is Sorting?

Sorting means arranging data in order.

Example:
```text
Before: [7, 2, 9, 1, 5]
After:  [1, 2, 5, 7, 9]
```

Sorting can also be done from largest to smallest.

# 2. What is Quick Sort?

Quick Sort chooses one element as a **pivot** and rearranges the data around it.

Example:
```text
[7, 2, 9, 1, 5]
```
Choose `5` as pivot:
```text
[2, 1] | 5 | [7, 9]
```
Everything smaller is placed on the left, and everything larger is placed on the right. Then the same idea is used on the two smaller parts.

# 3. What is a Pivot?

The pivot is the reference value used to divide the data.

Example:
```text
[8, 3, 5, 2, 7]
```
Choose `5`:
```text
smaller values | 5 | larger values
[3, 2]         | 5 | [8, 7]
```

The two sides do not have to be fully sorted yet.

# 4. Partitioning

Partitioning means rearranging values around the pivot.

Suppose:
```text
[8, 3, 5, 2, 7]
pivot = 5
```
After partitioning, a valid arrangement is:
```text
[3, 2] | 5 | [8, 7]
```

The important fact is that `5` is now in its final sorted position. Every value on the left is smaller, and every value on the right is larger.

# 5. Real-Life Idea

Imagine students standing in a line. Choose one student as the pivot, for example a student who is 170 cm tall.

Separate everyone into:
```text
Shorter than 170 | 170 | Taller than 170
```

You have not fully sorted the students, but you have divided the problem into smaller groups. This is partitioning.

# 6. Quick Sort Step by Step

Start:
```text
[4, 2, 7, 1, 5, 3]
```
Choose `3` as pivot. Conceptually:
```text
[2, 1] | 3 | [4, 7, 5]
```
Then sort the left and right groups:
```text
[1, 2] | 3 | [4, 5, 7]
```
Final:
```text
[1, 2, 3, 4, 5, 7]
```

# 7. Recursive Idea

After partitioning:
```text
Left | Pivot | Right
```

The pivot is already where it belongs. Quick Sort now applies the same process to the left and right sides.

So the pattern is:
```text
Choose pivot
↓
Partition
↓
Sort left
↓
Sort right
```

# 8. Pivot Choice

A pivot can be chosen in several ways:
- first element
- last element
- middle element
- random element

Good pivots tend to make balanced partitions:
```text
50 | pivot | 49
```
Bad pivots can create very unbalanced partitions:
```text
99 | pivot | 0
```

# 9. Average and Worst Case

If partitions are reasonably balanced, Quick Sort has about `log n` levels and each level processes about `n` elements:
```text
O(n) × O(log n) = O(n log n)
```

Average time:
```text
O(n log n)
```

If the pivot repeatedly creates one huge side and one tiny side, the work becomes:
```text
n + (n-1) + (n-2) + ...
```
which gives worst-case:
```text
O(n²)
```

# 10. Quick Sort vs Merge Sort

Merge Sort:
```text
Split → Sort both halves → Merge
```
Worst-case time: `O(n log n)`

Quick Sort:
```text
Pivot → Partition → Sort both sides
```
Average time: `O(n log n)`
Worst-case: `O(n²)`

# 11. Simple Quick Sort Code

This version is for learning because it creates separate lists for values smaller than, equal to, and larger than the pivot.

````python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

nums = [8, 3, 5, 4, 7, 6, 1, 2]
print("Original:", nums)
print("Sorted:  ", quick_sort(nums))
````

# 12. LeetCode 215 — Kth Largest Element

The problem gives an array and a number `k`. We need the kth largest element.

Example:
```text
nums = [3, 2, 1, 5, 6, 4]
k = 2
```
Sorted order:
```text
[1, 2, 3, 4, 5, 6]
```
The 2nd largest element is `5`.

# 13. Why We Do Not Need to Fully Sort

If a problem asks for the kth largest value, we only need one position.

For the example above, we do not really need to know the exact order of every element. We only need to identify the value that belongs at the 2nd-largest position.

This leads to **Quickselect**.

# 14. What is Quickselect?

Quickselect uses the same partition operation as Quick Sort. The difference is what happens after partitioning.

Quick Sort:
```text
Partition → work on left AND right
```

Quickselect:
```text
Partition → work only on the side containing the answer
```

This is the key optimization.

# 15. Convert Kth Largest to an Index

Python uses zero-based indexes. If there are `n` elements, then:
```text
Largest     → index n-1
2nd largest → index n-2
3rd largest → index n-3
```

Therefore:
```text
target index = n - k
```

Example:
```text
n = 6
k = 2
target = 6 - 2 = 4
```
In sorted order, index 4 contains `5`.

# 16. Quickselect Example

Start with:
```text
[3, 2, 1, 5, 6, 4]
```
Target index for `k = 2`:
```text
4
```
Suppose the pivot is `4`. After partitioning, `4` ends around index 3:
```text
[3, 2, 1] | 4 | [5, 6]
```
Our target is index 4, so the answer must be on the right. We can ignore the left side completely and continue with the right side.

# 17. Quickselect Decision Rule

After partitioning, suppose the pivot lands at `p` and our target index is `target`.

```text
p == target → answer found
p < target   → search right side
p > target   → search left side
```

Only one side continues. This is what makes Quickselect different from Quick Sort.

# 18. Duplicates

Duplicates count as separate elements.

Example:
```text
[5, 5, 4, 3]
```
Sorted:
```text
[3, 4, 5, 5]
```

For `k = 2`, the 2nd largest is still `5`. We count positions, not distinct values.

# 19. LeetCode 215 Solution

This solution uses in-place partitioning and Quickselect.

````python
class Solution:
    def findKthLargest(self, nums, k):
        target = len(nums) - k

        left = 0
        right = len(nums) - 1

        while left <= right:
            pivot_index = self.partition(nums, left, right)

            if pivot_index == target:
                return nums[pivot_index]

            if pivot_index < target:
                left = pivot_index + 1
            else:
                right = pivot_index - 1

        return -1

    def partition(self, nums, left, right):
        pivot = nums[right]
        store = left

        for i in range(left, right):
            if nums[i] < pivot:
                nums[store], nums[i] = nums[i], nums[store]
                store += 1

        nums[store], nums[right] = nums[right], nums[store]
        return store
````

# 20. Understanding `target = n - k`

This converts a kth-largest question into a normal ascending index.

Example:
```text
nums length = 9
k = 4
target = 9 - 4 = 5
```
The value at index 5 in ascending order is the 4th largest.

# 21. Understanding Partition in the Solution

The partition function uses the last value as the pivot. It moves values smaller than the pivot toward the left. At the end, it places the pivot after all smaller values.

The returned index is the pivot's final sorted position.

# 22. What Does `store` Mean?

`store` marks the next position where a value smaller than the pivot should be placed.

Example with pivot `4`:
```text
[3, 2, 1, 5, 6, 4]
```
Values `3`, `2`, and `1` are smaller than `4`, so they are moved to the left region. Then `4` is placed immediately after them.

# 23. Dry Run

Input:
```text
nums = [3, 2, 1, 5, 6, 4]
k = 2
```
Target index:
```text
6 - 2 = 4
```
Choose pivot `4`. Partitioning gives the pivot a position around index 3.

Because:
```text
3 < 4 target
```
we search the right side. That side contains the larger values. Continuing the same process places `5` at the target index.

Answer:
```text
5
```

# 26. Complexity of Quickselect

Average time:
```text
O(n)
```

After every partition, only one side is kept. Large parts of the array can be discarded without being fully sorted.

Worst case:
```text
O(n²)
```
when pivots are repeatedly very poor and the partition is extremely unbalanced.

# 27. Quick Sort vs Quickselect

Quick Sort goal:
```text
Sort everything
```
Average: `O(n log n)`
Worst: `O(n²)`

Quickselect goal:
```text
Find one target position
```
Average: `O(n)`
Worst: `O(n²)`

# 28. Common Mistakes

1. Confusing kth largest with kth smallest. Use `target = n - k`.

2. Forgetting duplicates. Equal values still occupy separate positions.

3. Continuing into both sides after partition. Quickselect should continue only into the side containing the target.

4. Assuming Quick Sort is always O(n log n). Its worst case is O(n²).

5. Forgetting that the problem asks for the value, not the index.

# 29. Final Memory Trick

Remember Quick Sort:
```text
PIVOT → PARTITION → SORT BOTH SIDES
```

Remember Quickselect:
```text
PIVOT → PARTITION → KEEP ONE SIDE
```

For LeetCode 215:
```text
kth largest
↓
target index = n - k
↓
partition
↓
compare pivot position with target
↓
search only the needed side
```

Most important complexities:
```text
Quick Sort average: O(n log n)
Quick Sort worst:   O(n²)
Quickselect average: O(n)
Quickselect worst:   O(n²)
```

The central idea is simple: **do not sort everything when the problem only asks for one position.**
